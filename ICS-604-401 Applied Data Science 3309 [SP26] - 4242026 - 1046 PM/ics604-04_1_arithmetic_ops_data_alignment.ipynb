## Vectorization

Vectorization refers to applying an operation to an entire collection of elements at once, rather than processing each element individually. This approach is implemented in the NumPy library and naturally extends to Pandas, which relies on NumPy arrays for data storage and computation.

By leveraging the parallel processing capabilities of modern CPUs, vectorized operations are significantly more efficient than traditional Python `for` loops. Instead of explicitly iterating over elements, vectorization performs operations on whole arrays at once, reducing overhead and improving performance.

A Pandas Series typically contains homogeneous data types (even though it can technically hold mixed types), whereas Python lists often contain heterogeneous elements. This consistency allows operations on Series to be executed by optimized, compiled C code. As a result, vectorized operations offer substantial performance improvements over equivalent Python code, which must process elements one by one and repeatedly check data types during execution.


````python
x = pd.Series(range(1_000_000))
display(x.head())
x.shape
````

````python
%%time
y = x**2
````

````python
%%time
a = range(1_000_000)
b = []
for elem in a:
    b.append(elem**2)
````

## The Numerical Python (NumPy) Library

NumPy is a core library for array-based computing in Python and is implemented primarily in the C programming language. It is widely regarded as the de facto standard for scientific computing and serves as the foundation of the Python scientific ecosystem, including libraries such as Pandas, SciPy, and Matplotlib. Pandas relies internally on NumPy arrays to store and manipulate data. These arrays are designed to store densely packed, homogeneous values, enabling efficient processing by low-level, optimized code.

NumPy arrays store data contiguously in memory, which allows for efficient use of the CPU cache and significantly faster data traversal. This memory layout is particularly advantageous when working with large numerical datasets, where both computational performance and memory efficiency are critical.

In addition to its efficient data representation, NumPy provides a comprehensive set of functions and methods for operating on arrays. These tools support fast numerical computations, transformations, and data analysis tasks, making NumPy an essential component of high-performance scientific and numerical computing in Python.

### Broadcasting

Broadcasting refers to a set of rules that allow NumPy to apply binary operations — such as addition, subtraction, or multiplication — to arrays of different shapes. Instead of requiring arrays to have exactly the same size, broadcasting determines how these arrays can interact in a well-defined and efficient way.

When performing operations, NumPy follows strict rules to align the shapes of the arrays involved. In the one-dimensional case, these rules are relatively straightforward. If the shapes of the two arrays do not match, NumPy checks whether one of the arrays has a size of 1. If so, that array is conceptually “stretched” to match the size of the other array.

However, if the array sizes differ and neither array has a dimension of size 1, broadcasting cannot be applied and NumPy raises an error. These rules ensure that broadcasting is both predictable and safe, while still enabling concise and efficient numerical computations.

````python
x = np.array([1, 2, 3])
y = np.array([3, 2, 1])

print(x.shape, y.shape)
x + y
````

````python
x = np.array([1, 2, 3])
y = np.array([1])

print(x.shape, y.shape)
x + y
````

````python
# Second condition throws and error
x = np.array([1, 2, 3])
y = np.array([1, 2])

print(x.shape, y.shape)
x + y
````

## Vectorization: Arithmetic Operations

Arithmetic operations in Pandas are vectorized, meaning they are applied to entire columns or Series at once rather than element by element in a loop. For example, you can add or multiply two columns directly, and Pandas will automatically perform the operation in parallel across all values.

To illustrate this, consider a simple DataFrame:


````python
df_1 = pd.DataFrame({'AA':{'A':79, 'C':2, 'T':13, 'X':21},
                     'BB':{'A':11, 'C':2, 'T':2, 'X':9}})
df_1
````

Adding two columns is straightforward and fully vectorized:

````python
df_1["AA"] + df_1["BB"]
````

Vectorized arithmetic in Pandas is not limited to operations on columns with identical indices or even Series of the same size. Pandas automatically aligns data based on index labels before performing the operation. For example, consider a second DataFrame:

````python
df_2 = pd.DataFrame({'AA':{'A':21, 'D':14, 'T':5},
                     'CC':{'A':12, 'D':28, 'T':121}})
df_2
````

````python
df_1["AA"] + df_2["AA"] 
````

When performing arithmetic operations between Series or DataFrames, Pandas matches values by their index labels. If a label exists in one object but not the other, the result will contain a missing value (`NaN`) for that label. This label-based alignment allows for flexible and expressive arithmetic while still benefiting from vectorized performance.

````python
df_1 + df_2
````

### Vectorized Arithmetic Between Series (DataFrame Columns/Rows)

Vectorized arithmetic operations between Pandas Series are performed by first aligning the data on their index labels. When two Series are involved in an operation, Pandas creates a new index that is the union of the indices from both Series. If a label is present in one Series but missing in the other, the result for that label is set to a missing value (`NaN`).

This alignment behavior applies not only to column-wise operations, but also to row-wise operations. Rows selected from a DataFrame using `.loc` are returned as Series, and arithmetic between them follows the same index-alignment rules. 

For example:

```python
df_1.loc["A"] + df_2.loc["D"]
```

````python
df_1.loc["A"] + df_2.loc["D"]
````

### Vectorized Arithmetic on DataFrames

Pandas also supports vectorized arithmetic operations on entire DataFrames. This behavior is a natural extension of the alignment rules used for Series operations. When performing arithmetic between two DataFrames, Pandas aligns both the row indices and column labels before applying the operation.

For example:

In this operation, Pandas constructs the union of the row indices from both DataFrames. Arithmetic is performed only where a row index is present in both DataFrames; otherwise, the result contains missing values (`NaN`).

````python
df_1 = pd.DataFrame({'AA':{'A':79, 'C':2, 'T':12, 'X':21},
                     'BB':{'A':11, 'C':2, 'T':2, 'X':9}})
df_2 = pd.DataFrame({'AA':{'A':21, 'D':14, 'T':5},
                     'BB':{'A':12, 'D':28, 'T':121}})
df_1 + df_2
````

### Broadcasting

When an arithmetic operation involves a Pandas Series and a single value (a *scalar*), Pandas applies the same broadcasting rules used in NumPy. The same logic also applies when performing operations between a DataFrame and a scalar.

In this case, the scalar value is conceptually expanded (broadcasted) to match the shape of the Series or DataFrame. The operation is then applied element-wise in a vectorized manner. For example:
  
```python
df_1['AA'] + 1.2
```

<img src="https://www.dropbox.com/scl/fi/0ppm09666cjdcy6n1cssk/alignment.png?rlkey=vo5bmcufx402h5e1z8uspb56i&st=0pi9adwt&dl=1" style="width:500px">

Here, the value `1.2` is added to every element in the `AA` column (a Series), producing a new Series where each value has been increased by the scalar amount.

````python
df_1 = pd.DataFrame({'AA':{'A':79, 'C':2, 'T':13, 'X':21},
                     'BB':{'A':11, 'C':2, 'T':2, 'X':9}})
df_1['AA'] + 1.2
````

Arithmetic operations between a DataFrame and a Series follow a well-defined alignment and broadcasting rule. When a Series is combined with a DataFrame, Pandas first aligns the Series index with one of the DataFrame’s axes — either the columns or the rows — based on matching labels. If the Series index matches the column labels of the DataFrame, the operation is applied column-wise, broadcasting the corresponding Series values down each column.

In the example below, the Series `s` has an index that matches the column labels of the DataFrame `df_1`:

````python
s = pd.Series([100, 200], index=['AA', 'BB'])
df_1 + s
````

Before performing the addition, Pandas aligns the Series values to the corresponding DataFrame columns. The value `100` is aligned with column `AA`, and `200` is aligned with column `BB`. Once aligned, these values are broadcast down each column, meaning that the same value is added to every element within the matched column.

As a result, `100` is added to all values in column `AA`, and `200` is added to all values in column `BB`. This broadcasting occurs without explicitly looping over rows or columns and is handled internally by optimized, vectorized code. The combination of label-based alignment and automatic broadcasting allows Pandas to perform such operations efficiently while preserving clarity and correctness in the presence of labeled data.

````python
# Align s with the row labels of df_1 by specifying the axis 
s = pd.Series([100, 200], index=['A', 'C'])
df_1.add(s, axis=0)
````

````python
s = pd.Series([100, 200, 300], index=['A', 'B', 'C'])
df_1.add(s, axis=0)
````

## Conditional Subsetting

In data analysis, it is often necessary to work with only a portion of a dataset. We already know how to subset a DataFrame to extract a single row using indexing. For example:

```python 
spending_df.iloc[1]
```
We also know how to extract a range of rows:

```python 
spending_df.iloc[1:32]
```

Beyond indexing, it is often even more useful to subset data based on **logical conditions**. This allows us to answer practical, real-world questions, such as extracting all entries related to cardiology, counting how many flights departing from HNL were headed to Nebraska, or determining how many Biki bicycles were checked out within a specific geographical region. Conditional subsetting enables focused analysis by selecting only the subset of data that meets specific criteria.

### Comparison Operations

Comparison operators such as (`<`, `>`, `==`, `>=`, `<=`, `!=`) work in Pandas much like arithmetic operations, but instead of producing numeric results, they return Boolean values (`True` or `False`). For example, comparing the `AA` column in two DataFrames:

```python
>>> df_1.loc[:, "AA"] > df_3.loc[:, "AA"]
A     True
C    False
T     True
X    False
Name: AA, dtype: bool
```

produces a Series of Booleans with the same indices as the original DataFrames. It is important to note that comparisons between two Series or two DataFrames can only be performed if both objects are **identically labeled**, meaning they must have the same row **and** column names. All entries in the resulting Series or DataFrame are of type `bool`: `True` indicates that the comparison test passed, while `False` indicates it did not.

If the labels do not match, as in this example:

```python
>>> df_1.loc[:, "AA"] > df_2.loc[:, "AA"]
...
ValueError: Can only compare identically-labeled Series objects
```

Pandas will raise an error, emphasizing the importance of proper alignment for comparison operations.

````python
df_1 = pd.DataFrame({'AA':{'A':79, 'C':2, 'T':13, 'X':21},
                     'BB':{'A':11, 'C':2, 'T':2, 'X':9}})
df_3 = pd.DataFrame({'AA':{'A':64, 'C':2, 'T':6, 'X':22},
                     'BB':{'A':3,  'C':2, 'T':7, 'X':12}})

df_1["AA"] > df_3["AA"]
````

````python
## Not the same number of rows nor the same index

df_4 = pd.DataFrame({'AA':{'I':64, 'J':2, 'K':6}})

df_1["AA"] > df_4["AA"]
````

### Boolean Operations

Conditional expressions in Pandas allow us to create more sophisticated queries by combining multiple logical conditions. These queries commonly rely on Boolean logic, but Pandas uses slightly different operators than standard Python. Specifically, `and` is replaced by `&`, `or` is replaced by `|`, and `not` is replaced by `~`. These operators return a Boolean object (Series or DataFrame) of the same shape as the input, which can then be used for subsetting or further analysis.

For example, to find all instances in `df_1` where the values in both columns `'AA'` and `'BB'` are greater than 3, we can use:

```python
>>> true_false_series = (df_1['AA'] > 3) & (df_1['BB'] > 3) 
```

The resulting Series of Booleans indicates which rows satisfy both conditions simultaneously, enabling precise filtering of the dataset.

````python
true_false_series = (df_1['AA'] > 3) & (df_1['BB'] > 3) 
true_false_series
````

````python
type(true_false_series)
````

````python
df_1[true_false_series]
````

````python
df_1[(df_1['AA'] > 3) & (df_1['BB'] > 3)]
````

### Boolean Operations Examples with the Medical Spending Data Set

Suppose we want to extract rows from `spending_df` where the specialty is `"DENTIST"` and the `spending`is below $100. We can start by creating individual Boolean Series for each condition:

- Condition 1: `specialty` is `'DENTIST'`

```python
>>> spending_df['specialty'] == 'DENTIST'
```

````python
(spending_df['specialty'] == 'DENTIST').head()
````

- Condition 2: `spending` below $100

```python 
>>> spending_df['spending'] < 100
``` 

````python
(spending_df['spending'] < 100).head()
````

- We can then combine these conditions using the `&` operator to capture rows that satisfy both criteria:

```python
(spending_df['specialty'] == 'DENTIST') & (spending_df['spending'] < 100)
```

````python
((spending_df['specialty'] == 'DENTIST') & (spending_df['spending'] < 100)).head()
````

- Finally, we use this Boolean Series to subset the original DataFrame and save the results to a new DataFrame:

```python
dentist_and_small_spending = (spending_df['specialty'] == 'DENTIST') & (spending_df['spending'] < 100)
spending_df.loc[dentist_and_small_spending]
```

````python
dentist_and_small_spending = (spending_df['specialty'] == 'DENTIST') & (spending_df['spending'] < 100)
spending_df.loc[dentist_and_small_spending].head()
````

The resulting DataFrame contains only those rows where the `specialty` is `"DENTIST"` and the `spending` is under $100, demonstrating how Boolean operations enable precise, multi-condition filtering.
