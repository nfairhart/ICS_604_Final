## Regression Analysis

The primary goal of regression analysis is to determine whether one or more predictor variables $X$ (or $X_1, X_2,\dots, X_p$) are associated with an outcome variable $Y$, and if so, to describe the nature of that relationship and use it for prediction. In many applications, we begin with data where the outcome is known and use it to train a model that can later be applied to new data where the outcome is unknown. This process is known as supervised learning. Importantly, regression serves a dual purpose: it can be used both to **predict outcomes** and to **explain relationships** between variables.

Historically, regression has been widely used as a tool for explanation, particularly in scientific and social science disciplines. The primary goal in these settings is often to understand how changes in predictor variables influence an outcome. For example, economists may study how consumer spending affects GDP growth, while business managers — such as those at ABC stores — may want to understand whether increased investment in tourism promotion leads to higher sales. In these contexts, the key quantity of interest is the strength and direction of the relationship. A major advantage of regression here is that it provides an interpretable, transparent model, unlike more complex “black box” methods.

In contrast, in machine learning and many applied fields, the emphasis shifts toward prediction rather than explanation. The goal is to build a model that can accurately estimate outcomes for new, unseen data. In this case, the primary focus is on the fitted values $\hat{Y}$, which represent the model’s predictions. For instance, marketers may use regression to predict how revenue will change in response to different advertising budgets, while universities might use it to estimate students’ future GPA based on standardized test scores. These applications prioritize predictive accuracy over interpretability, though both perspectives are valuable depending on the problem at hand.

For a deeper discussion of the distinction between explanation and prediction, see the paper [To Explain or to Predict?](https://projecteuclid.org/journals/statistical-science/volume-25/issue-3/To-Explain-or-to-Predict/10.1214/10-STS330.full), which highlights how the goals of modeling can influence the choice of methods and evaluation criteria.

### Simple Linear Regression

Simple linear regression (also called single linear regression) is one of the most fundamental tools for modeling relationships between variables. It describes how a single predictor variable $X$ is related to a quantitative response variable $Y$. The method assumes that this relationship is approximately linear, meaning that as $X$ increases, $Y$ either increases or decreases in a consistent way. While correlation tells us how strongly two variables are associated, regression goes a step further by providing an explicit mathematical model that describes how $Y$ changes as a function of $X$. This relationship is expressed through the equation of a line:

$$
    Y = \beta_0 + \beta_1X
$$

where $\beta_0$ is the intercept and $\beta_1$ is the slope.

Several key terms are essential for understanding regression. The **response variable** (or dependent variable, target, outcome) is the variable we aim to predict, typically denoted $Y$. The **independent variable** (or predictor, feature, attribute) is the input variable $X$ used to make predictions. The **intercept** $\beta_0$ represents the predicted value of $Y$ when $X=0$, while the **regression coefficient** $\beta_1$ (slope) indicates how much $Y$ changes for a one-unit increase in $X$. The **fitted values** are the predicted values $\hat{Y}_i$ obtained from the regression line, and the **residuals** are the differences between the observed and predicted values. Finally, the model is typically fit using the **least squares** method, which finds the line that minimizes the sum of squared residuals, ensuring the best overall fit to the data.

### Example: Advertising Data

This example illustrates how regression can be used to analyze the relationship between advertising spending and product sales. The dataset contains sales (measured in thousands of units) and advertising budgets (in thousands of dollars) across three media channels: TV, radio, and newspaper. A key objective is to determine which advertising medium contributes most to sales, how precisely we can estimate these effects, and whether combinations of media produce synergistic effects beyond their individual contributions.

````python
advertising = pd.read_csv('data/Advertising.csv', usecols=[1, 2, 3, 4])
advertising.head()
````

````python
from scipy.stats import linregress

linreg = linregress(advertising["TV"], advertising["sales"])
linreg
````

<br>
Using a simple linear regression model, we focus on the relationship between TV advertising and sales. The `linregress` function estimates the intercept $\beta_0$ and slope $\beta_1$, giving us the model:

$$
    sales \approx \beta_0 + \beta_1 \times \text{TV}
$$

Here, $\beta_0$ represents the baseline level of sales when TV advertising is zero, and $\beta_1$ represents the expected increase in sales for each additional unit of TV advertising spending. These two values together form the **model parameters**, which define the fitted regression line. Once estimated, the model can be used to make predictions — for example, by plugging in a TV advertising budget of 250 (i.e., $250,000), we obtain a predicted sales value using the regression equation.

````python
some_tv_advertising = 250
some_sales_prediction = linreg.intercept + linreg.slope * some_tv_advertising

print(some_sales_prediction)
````

````python
some_tv_advertising = 250
some_sales_prediction = linreg.intercept + linreg.slope * some_tv_advertising

plt.figure(figsize=(10, 4))
plt.scatter(advertising["TV"], advertising["sales"], alpha=0.5)
plt.xlim(-10, 320)
plt.ylim(0, 28)
plt.vlines(some_tv_advertising, 0, some_sales_prediction, color='red')
plt.hlines(some_sales_prediction, -10, some_tv_advertising, color='red')
plt.scatter(some_tv_advertising, some_sales_prediction, s=50, color='red');
````

````python
x_axis = np.arange(0, 320)
y_line = linreg.intercept + linreg.slope * x_axis

plt.figure(figsize=(10, 4))
plt.scatter(advertising["TV"], advertising["sales"], alpha=0.5)
plt.plot(x_axis, y_line, color='red', linewidth=4, alpha=0.5);
````

### The Residual Sum of Squares

The residual sum of squares (RSS) is a key measure used to quantify how well a regression line fits the data. Each residual $e_i$ represents the difference between the predicted value and the observed value for a data point. Since residuals can be positive or negative, simply summing them could lead to cancellation. To avoid this, we square each residual and sum them:

$$
RSS = e_1^2 + e_2^2 + \cdots + e_n^2
$$

A smaller RSS indicates that the model’s predictions are closer to the observed data, meaning a better fit.

### Minimizing the RSS

In linear regression, our goal is to find the model parameters $\beta_0$ (intercept) and $\beta_1$ (slope) that minimize the RSS. The line that achieves the smallest RSS is considered the best-fitting line under the assumption of a linear relationship. This is why linear regression is often referred to as a **least squares** method — it selects the parameters that minimize the sum of squared residuals.

````python
plt.figure(figsize=(16, 10))

x_axis = np.arange(-10, 320)
models = [(linreg.intercept, linreg.slope), (-2, 0.2), (20, 0), (20, -0.1)]

for (m, model) in enumerate(models):
    plt.subplot(2, 2, m+1)
    y_line = model[0] + model[1] * x_axis
    plt.scatter(advertising["TV"], advertising["sales"], alpha=0.2)
    RSS = 0 
    for i in range(advertising.shape[0]):
        y_hat = model[0] + model[1] * advertising.iloc[i]["TV"]
        plt.vlines(advertising.iloc[i]["TV"], advertising.iloc[i]["sales"], y_hat, color='red')    
        RSS += (y_hat - advertising.iloc[i]["sales"])**2
    plt.plot(x_axis, y_line)
    plt.title("RSS={}".format(round(RSS, 1)), fontsize=16)
````

<br>

For simple linear regression, we can derive exact formulas for these parameters using calculus. By taking partial derivatives of the RSS with respect to $\beta_0$ and $\beta_1$, setting them equal to zero, and solving, we obtain:

$
\hspace{2in} \beta_1 = \frac{\sum_{i=1}^{n} (x_i -\bar{x})(y_i -\bar{y})}{\sum_{i=1}^{n} (x_i -\bar{x})^2}
$

$
\hspace{2in} \beta_0 = \bar{y} - \beta_1\bar{x}
$

These closed-form solutions make linear regression computationally efficient and easy to interpret. In more complex machine learning models, however, such closed-form solutions may not exist, and we instead rely on iterative optimization techniques such as gradient descent to find parameters that minimize the error.

An important consideration is that our dataset is typically just a sample from a larger population. If we were to collect a different sample, we might obtain slightly different estimates for 
$\beta_0$ and $\beta_1$. This raises the question of **how confident we are in our estimated parameters**. To address this uncertainty, we can construct **confidence intervals**, which provide a range of plausible values for the true parameters.

One practical way to estimate these confidence intervals is through **bootstrapping**, a method we have previously discussed. By repeatedly resampling the data (with replacement), fitting a regression model to each resampled dataset, and collecting the resulting parameter estimates, we can build a distribution for 
$\beta_0$ and $\beta_1$. From this distribution, we can compute a 95% confidence interval, giving us a sense of how stable and reliable our model estimates are.

````python
print(advertising.shape)
````

````python
# bootstrapping the data (same sample size)
advertising.sample(advertising.shape[0], replace=True).head()
````

````python
advertising.sample(advertising.shape[0], replace=True).head()
````

````python
bootstrap_intercepts = []
bootstrap_slopes = []

for i in range(50_000):
    sample_indexes = np.random.choice(advertising.index, size=len(advertising.index))
    bootstrap_sample = advertising.iloc[sample_indexes]
    lr = linregress(bootstrap_sample["TV"], bootstrap_sample["sales"])    
    bootstrap_intercepts.append(round(lr.intercept, 3))
    bootstrap_slopes.append(round(lr.slope, 3))

print("10 first intercepts:")    
print(bootstrap_intercepts[:10])
print("10 first slopes:")    
print(bootstrap_slopes[:10])
````

````python
plt.figure(figsize=(10, 4))
plt.scatter(advertising["TV"], advertising["sales"], alpha=0.2)

x_axis = np.arange(-10, 320)

for i in np.random.choice(range(50000), 1000, replace=False):
    y_hat = bootstrap_intercepts[i] + bootstrap_slopes[i] * x_axis
    plt.plot(x_axis, y_hat, color='red', alpha=0.02)
````

````python
plt.figure(figsize=(8, 4))

plt.hist(bootstrap_intercepts, edgecolor='black', linewidth=1.2)
plt.title("Distribution of Intercepts")

conf_interval = np.percentile(bootstrap_intercepts, [2.5, 97.5])
print(conf_interval)

plt.axvline(conf_interval[0], color="red")
plt.axvline(conf_interval[1], color="red");
````

````python
plt.figure(figsize=(8, 4))

plt.hist(bootstrap_slopes, edgecolor='black', linewidth=1.2)
plt.title("Distribution of Slopes")

conf_interval = np.percentile(bootstrap_slopes, [2.5, 97.5])
print(conf_interval)

plt.axvline(conf_interval[0], color="red")
plt.axvline(conf_interval[1], color="red");
````

### Statsmodels

The `Statsmodels` package in Python provides a powerful and comprehensive framework for performing statistical modeling, including linear regression. In particular, it allows us to fit models using *ordinary least squares (OLS)*, which is the standard method for estimating the parameters of a linear regression model. As discussed earlier, OLS works by finding the values of the model parameters that minimize the residual sum of squares (RSS), ensuring the best possible fit under the assumption of a linear relationship.

One of the convenient features of Statsmodels is its use of a formula-based syntax inspired by the statistical programming language R. In this notation, we describe models using expressions like:

$$
\text{sales} \sim TV,
$$

where the symbol “$\sim$” can be read as “is a function of” or “depends on.” This means we are modeling *sales* as a function of *TV advertising*. The formula interface makes it easy to specify models with multiple predictors as well, such as including radio and newspaper advertising in addition to TV.

Beyond simply estimating coefficients, Statsmodels provides rich output, including standard errors, confidence intervals, hypothesis tests, and goodness-of-fit measures. This makes it especially useful when the goal is not just prediction, but also statistical inference — understanding the reliability and significance of the relationships captured by the model.

````python
linregress(advertising["TV"], advertising["sales"])
````

````python
import statsmodels.formula.api as smf

est = smf.ols('sales ~ TV', advertising).fit()
est.summary().tables[1]
````

````python
est.params
````

````python
# compute predicted values given the estimated model parameters

est.predict(advertising["TV"])
````

### Interpreting the Simple Linear Regression

Interpreting the parameters of a simple linear regression model helps us understand both the baseline behavior of the response variable and how it changes with the predictor.

The intercept, $\beta_0=7.0326$, represents the expected value of sales when the TV advertising budget is zero. In this context, it suggests that if no money is spent on TV advertising, the model predicts average sales of approximately 7.03 thousand units. However, the intercept does not always have a meaningful real-world interpretation. This is especially true when a value of $x=0$ is unrealistic or outside the range of the observed data. For example, if all observed advertising budgets are well above zero, then interpreting the intercept as a real scenario may not make practical sense — it is simply a mathematical artifact needed to define the regression line.

The slope, $\beta_1=0.0475$, is often the more informative parameter. It quantifies the association between TV advertising and sales by measuring the average change in sales for a one-unit increase in the predictor. Since the units here are in thousands of dollars and thousands of units sold, this means that for every additional $1,000 spent on TV advertising, sales increase by approximately 0.0475 thousand units, or about 47.5 units. This provides a clear and actionable interpretation: increased investment in TV advertising is associated with higher sales, and the slope tells us the expected magnitude of that increase on average.

### Multiple Linear Regression

When predicting sales, it is often beneficial to incorporate multiple sources of information rather than relying on a single variable. A naive approach might be to build separate regression models for each predictor (e.g., TV, radio, newspaper) and then attempt to combine their predictions. However, this approach has clear limitations. It is not obvious how to combine the results in a principled way (e.g., should we take the mean, minimum, or median?), and more importantly, each model ignores the influence of the other variables. This means it fails to capture how predictors may interact or overlap in their effects on the response.

````python
est = smf.ols('sales ~ radio', advertising).fit()
est.summary().tables[1]
````

````python
est = smf.ols('sales ~ newspaper', advertising).fit()
est.summary().tables[1]
````

<br>

To address these issues, we use **multiple linear regression**, which extends simple linear regression by incorporating multiple predictors into a single model:

$$
y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \cdots + \beta_p x_p + \epsilon.
$$

Here, $p$ is the number of predictors, and each coefficient $\beta_i$ represents the average effect on $y$ of a one-unit increase in $x_i$, *holding all other variables constant*. This “holding others constant” interpretation is crucial — it allows us to isolate the unique contribution of each predictor while accounting for the presence of the others.

In the advertising example, this leads to a model of the form:

$$
sales = \beta_0 + \beta_1 \times \text{TV\_budget} + \beta_2 \times \text{radio\_budget} + \beta_3 \times \text{newspaper\_budget} + \epsilon
$$

By fitting this model (e.g., using `statsmodels` and ordinary least squares), we can estimate how each advertising channel contributes to sales while controlling for the others. The output summary provides not only coefficient estimates, but also measures of uncertainty (such as standard errors), statistical significance (p-values), and overall model fit.

This approach allows us to answer more nuanced questions: Which advertising channels have the strongest independent effect on sales? Are some channels redundant when others are included? Do certain variables lose significance once we control for others? Multiple linear regression provides a unified and statistically sound framework for addressing these kinds of questions, making it far more powerful than combining separate simple models.

### Model Interpretation

Interpreting the results of a multiple linear regression model requires understanding that each coefficient reflects the effect of a predictor while holding all other variables constant. In this model, the coefficient for TV advertising indicates that for every additional $\$$1,000 spent on TV, sales are expected to increase by approximately 45.8 units on average, assuming radio and newspaper spending remain unchanged. Similarly, the coefficient for radio advertising suggests a much larger effect: each additional $1,000 spent on radio is associated with an average increase of about 188.5 units in sales, again holding the other variables fixed. This indicates that, in this model, radio advertising has a stronger marginal impact on sales than TV advertising.

In contrast, newspaper advertising is found to be statistically insignificant, meaning its coefficient is not reliably different from zero once we account for TV and radio spending. This indicates that, after controlling for the effects of these other variables, newspaper advertising does not add meaningful explanatory power for predicting sales. However, this raises an important question: why does newspaper advertising appear to have a positive effect when analyzed on its own using simple linear regression, where it shows an estimated increase of about 54.7 units per $1,000, but not in the multiple regression model?

The answer lies in the **correlation between predictors**. Radio and newspaper advertising are moderately positively correlated (around 0.35), meaning that higher spending on radio is often accompanied by higher spending on newspaper. Since radio advertising has a strong and genuine effect on sales, some of that effect is inadvertently captured by newspaper advertising when it is considered in isolation. In other words, newspaper appears predictive in the simple regression model not because it directly drives sales, but because it is associated with another variable (radio) that does.

When we move to multiple linear regression, the model is able to disentangle these overlapping relationships. By including both radio and newspaper in the model, it can isolate the unique contribution of each variable. Because radio already explains much of the variation in sales, the additional contribution from newspaper becomes negligible and statistically insignificant. This demonstrates an important principle: relationships that appear meaningful in isolation can disappear once we control for other relevant variables. Multiple regression helps isolate the true, independent impact of each predictor by accounting for these interactions and correlations among variables.

### Some Important Concepts to Explore

As the number of predictors in a regression model grows, the problem of understanding and selecting the right variables becomes increasingly complex. With dozens (or even hundreds) of predictors, it is no longer practical to manually examine how each variable relates to others or to the response. Relationships between predictors — such as correlations and interactions — can become tangled, making it difficult to determine which variables truly matter. This complexity highlights the need for systematic approaches to model building and evaluation.

One important issue that arises in this setting is overfitting. As we add more predictors to a model, the $R^2$ value will almost always increase, because the model has more flexibility to fit the training data. However, this does not necessarily mean the model is better. In fact, it may simply be capturing noise rather than true underlying patterns, leading to poor performance on new, unseen data. Thus, a high $R^2$ alone is not a reliable indicator of model quality.

Selecting the best subset of predictors is therefore a key challenge. Ideally, we want a model that balances simplicity and predictive accuracy — capturing the important relationships without including unnecessary variables. This topic is often referred to as *model selection*, with *feature selection* being a more specific case focused on choosing the most relevant predictors. It is explored in depth in statistical learning literature, where various techniques are introduced to systematically evaluate and compare models.

To address these challenges, more advanced methods have been developed that automatically control model complexity. Techniques such as *ridge regression* and *lasso regression* introduce a form of regularization that “shrinks” the coefficients of less important variables toward zero. In particular, lasso regression can even reduce some coefficients exactly to zero, effectively performing variable selection. These approaches help prevent overfitting and produce models that are both more interpretable and better at generalizing to new data.
