## An Example of Probability in Data Science: A/B Testing

A/B testing is a statistical method widely used in data science to compare two alternatives and determine which one performs better with respect to a specific objective. The process begins by *randomly assigning* users to two groups (A and B) to ensure that any observed differences in outcomes are attributable to the treatment rather than pre-existing differences between users. Outcomes such as clicks, purchases, or conversions are then observed and modeled using appropriate *probability distributions*, depending on the nature of the data (for example, Bernoulli or Binomial models for conversion events).

From the collected sample data, we *estimate key parameters*, such as the conversion rates for each group. Using these estimates, we apply *hypothesis testing and confidence intervals* to assess whether the observed differences are likely due to chance or reflect a true underlying effect. This probabilistic framework allows organizations to *make data-driven decisions under uncertainty*, balancing evidence, risk, and statistical reliability when choosing between competing alternatives.

## Basic Concepts of Probability: Experiments, Sample Spaces, and Events

An `experiment` in probability is a procedure or random process that produces exactly one outcome from a set of possible outcomes. The key idea is uncertainty: although the process is well defined, the specific outcome cannot be predicted with certainty in advance. Examples include tossing a coin and observing whether it lands on heads or tails, rolling a die and recording the number shown, counting the number of empty slots at a Biki stand, or measuring the weight of a patient in a hospital.

The `sample space`, denoted by *S*, is the complete set of all possible, distinct outcomes of an experiment. For a coin toss, the sample space consists of {Heads, Tails}. When rolling a die, the sample space is the finite set {1, 2, 3, 4, 5, 6}. In other situations, the sample space may be larger or even infinite — for example, the number of empty slots at a Biki stand can vary over a range of integers, and a patient’s weight can take on any value within a continuous interval, such as from 0 up to some maximum value.

An `event`, denoted by *E*, is any subset of the sample space. Events describe outcomes or collections of outcomes we are interested in analyzing. Examples include the event that a coin lands on heads, a die shows a 4, or a die lands on an odd number (1, 3, or 5). In real-world settings, an event could be that there are no empty Biki slots available, or that a patient’s weight falls within a specific range, such as greater than 140 pounds but less than 200 pounds.

## Probability: Quantifying Uncertainty

Probability provides a way to quantify our beliefs abut uncertain events. It allows us to measure how likely or unlikely an event is to occur when the outcome is not known in advance. In data science and statistics, probability gives us a mathematical language for reasoning under uncertainty rather than relying on intuition alone.

Probabilistic reasoning appears in many real-world questions. For example, we may ask how likely a person is to default on a loan, donate to a fundraising campaign, or contract the flu during a given season. Similarly, probability can be used to assess operational and infrastructure questions, such as the likelihood that all Biki slots at a busy intersection are full, or that three traffic accidents occur on the H1 in a single day. 

We sometimes refer to a null probability (0) when an event is impossible, and a probability of 1 when an event is certain to occur. In general, a probability value is a number that quantifies the likelihood of an event and must satisfy two fundamental properties: it must lie between 0 and 1, and the probabilities of all possible outcomes in the sample space must sum to 1. That is, for any event $E$, 
$$
0 \leq p(E) \leq 1 \quad \text{and} \quad p(S) = 1
$$
where $S$ denotes the sample space. In the discrete case, this implies that the probabilities of all possible outcomes sum to 1, $\sum_{o \in S}{p(o) = 1}$. In the continuous case, probabilities are defined through a probability density function, and the total probability over the sample space is given by an integral, $\int_{S}f(x)dx = 1$. These properties ensure internal consistency, though a complete definition of probability also requires additivity over (countable) collections of disjoint events.

Importantly, not every number between 0 and 1 is a valid probability. A value can only be interpreted as a probability if it fits within a complete probability model in which all outcomes in the sample space together have total probability 1. In data analysis, probabilities are most often assigned using probability distributions. A probability distribution is a function that specifies the probability of each outcome — or event — in the sample space, providing a structured way to model uncertainty and make predictions.

## Random Samples

A **random sample** is a collection of observations selected in a way that allows probability theory to be applied. In a formal, mathematical setting, random samples can arise either from sampling a population or from repeating a random experiment under identical conditions.

- __Random sampling from a population.__
Let $P$ represent a population consisting of $N$ elements. A subset $R \subset P$ is called a random sample if each element $x \in P$ has a non-zero probability $p(x)$ of being included in $R$, where $0 \leq p(x) \leq 1$. For example, if $P$ represents all people living in a city, then $R$ could be a randomly selected group of residents. In this case, $p(x)$ represents the probability that any given individual is selected for the sample. 

- __Random sampling from repeated experiments.__
Random samples also arise from repeating the same experiment multiple times under identical conditions. Let $O = {o_1, o_2, ..., o_m}$ denote the set of outcomes from $m$ repetitions of an experiment. The collection $O$ forms a random sample of the experiment’s possible outcomes. For instance, repeatedly flipping a coin produces a sample of heads and tails outcomes. Similarly, recording the number of open slots at a Biki station over multiple days yields a sample of availability counts, and repeatedly measuring a patient’s weight produces a sample of weight observations.

Regardless of how it is obtained, a random sample has two key characteristics. **Randomness** ensures that each member of the population — or each outcome of the experiment — has a well-defined chance of being selected, helping to avoid systematic bias. **Independence** means that the selection of one observation does not affect the selection of another, allowing each observation to be treated as an independent piece of information. Together, these properties make random samples essential for valid statistical inference and data analysis.

## Long Run Frequencies of an Event

The concept of **long-run frequency** is a fundamental idea in probability. It reflects the tendency for outcomes to stabilize around their theoretical probabilities when an experiment is repeated many times. For example, in a casino game where the chance of winning is 1 in 10, playing ten times does not guarantee a win. Short-term outcomes can vary widely, but over a large number of plays, the observed frequency of winning tends to approach the theoretical probability of 0.1.

Consider flipping a fair coin. Intuitively, we expect that in the long run, the number of heads will be roughly equal to the number of tails, giving about 50% heads and 50% tails. Formally, the probabilities are $p(H)=0.5$ and $p(T)=0.5$. Although any single coin toss is unpredictable, the relative frequencies converge to these probabilities as the number of tosses increases.

This principle can be demonstrated computationally. For example, in Python, we can simulate repeated coin tosses using the `random` module to generate sequences of heads (1s) and tails (0s) and observe how the long-run frequencies approach the expected probabilities.

### Inspecting the Number of Accumulated Heads 

When simulating repeated coin tosses, it is often useful to track how the number of heads accumulates over time. This can be done using the cumulative sum, which adds up the values of a sequence progressively. Formally, for an array $X$, the cumulative sum at position $a$ (where $a \geq 1$) is calculated as:
$$
cumsum(X)[a] = \sum_{i=0}^a X[i]
$$

For a coin toss simulation, if we encode heads as 1 and tails as 0, the cumulative sum at each position gives the total number of heads observed up to that point.

In Python, the `cumsum()` function from the numpy package makes this calculation straightforward:

````python
random_trials = [random.choice([0, 1]) for _ in range(1000)]
cum_sum_random_trials = np.cumsum(random_trials)
    
print("-" * 30)
print(random_trials[0:10])

print("-" * 30)
print(cum_sum_random_trials[0:10].tolist())
````

````python
# Let's normalize `cum_sum_random_trials` by the number of trials in each position

cumulative_prop_heads = cum_sum_random_trials / range(1, len(cum_sum_random_trials)+1)

print("=====================")
print(random_trials[0:6])
print("---------------------")
print(cum_sum_random_trials[0:7].tolist())
print("---------------------")
print(cumulative_prop_heads[0:7].tolist())
print("=====================")
````

````python
plt.figure(figsize=(8, 4))
plt.plot(range(100), cumulative_prop_heads[0:100])
plt.axhline(0.5, color='k', linestyle='--', alpha=0.3);
````

````python
plt.figure(figsize=(12, 4))
plt.plot(range(len(cumulative_prop_heads)), cumulative_prop_heads)
plt.axhline(0.5, color='k', linestyle='--', alpha=0.3);
````

### Long-Run Frequency

Even after 1,000 coin flips, the proportion of heads in that set may not be exactly 0.5. A run of 1,000 flips is a random sample from the underlying probability distribution, so the relative frequency within that sample can differ from the theoretical probability.

However, as the number of trials increases, the long-run frequency tends to stabilize and approach the theoretical probabilities $p(H)$ and $p(T)$. This illustrates the *law of large numbers*: while individual outcomes are unpredictable — as they can deviate from the expected probability — the overall pattern becomes increasingly predictable as the number of repetitions grows. 

````python
random_trials = [random.choice([0, 1]) for _ in range(500_000)]

cum_sum_random_trials = np.cumsum(random_trials)
cumulative_prop_heads = cum_sum_random_trials / range(1, len(cum_sum_random_trials)+1)

plt.figure(figsize=(12, 4))
plt.plot(range(len(cumulative_prop_heads)), cumulative_prop_heads)
plt.axhline(0.5, color='k', linestyle='--', alpha=0.3);
````

## Probability versus Statistics

Probability and statistics are closely related fields, but they differ in focus and purpose. Probability is a theoretical branch of mathematics concerned with predicting the likelihood of future events based on known models or assumptions. It provides the tools to reason about uncertainty before observations are made.

In contrast, statistics is an applied branch of mathematics that seeks to make sense of real-world data. It analyzes and interprets the frequency and patterns of past events to draw inferences, estimate probabilities, or make decisions under uncertainty. A simple, but useful, distinction is that probability is primarily about predicting what might happen, while statistics is primarily about understanding what has already happened.

### The Underlying Generative Process

Populations are often dynamic, with new members continuously being added over time. To understand how these new members arise, we rely on the concept of a **generative process**, which summarizes the population by describing how new observations can be “realized.” The generative process specifies the probabilistic rules that govern how outcomes are produced, capturing the underlying mechanisms that give rise to the data.

This perspective naturally connects probability and statistics. Probability focuses on modeling the generative process itself — describing how data could be generated and assigning probabilities to possible outcomes. Statistics, on the other hand, works in the opposite direction: given observed data that are assumed to be generated by some underlying process, it seeks to infer properties of that process, estimate unknown parameters, and assess uncertainty. Together, probability and statistics provide a coherent framework for reasoning about how data are generated and how we learn from the data we observe.

### Using Simulation to Infer a Statistic for Generative Process

In many real-world problems, deriving probabilities analytically can be difficult or even infeasible. Closed-form solutions may require handling complex conditional structures or hierarchical dependencies — for example, flipping a coin to decide which die to use, then drawing from a distribution determined by that die. As the generative process becomes more intricate, the mathematics can quickly become unwieldy.

In such cases, **simulation** provides a practical alternative. By simulating the generative process a large number of times, we can approximate probabilities and other statistics of interest from the simulated outcomes. Rather than solving the problem symbolically, we rely on repeated sampling to estimate the behavior of the system.

A key challenge with simulation is determining whether the results have *converged* to the correct values. A small number of simulations — such as flipping a coin only ten times — can produce highly variable estimates that do not reflect the true underlying probabilities. Confidence in simulation-based results comes from running the experiment many times and observing stability in the estimated statistics as the number of simulations increases.

A classic example where simulation is especially useful is the Monty Hall problem, a probability puzzle based on the television game show *Let’s Make a Deal*, named after its original host, Monty Hall. The problem’s unintuitive outcome makes it an ideal illustration of how simulation can reveal the behavior of a generative process when analytical reasoning alone can be misleading.

### The Monty Hall Problem
 
```
Suppose you are a contestant on a game show and are presented with three doors. Behind one door is a car (the prize) and behind the other two doors are goats. You select one door — say Door 1 — but do not open it. The host, who knows what is behind each door, then opens one of the two remaining doors, say Door 3, revealing a goat. The host then offers you a choice: remain with your original selection or switch to the other unopened door." 
```
The central question is whether switching doors improves your chances of winning the car. Although the situation may appear symmetric after one door is opened, the host’s action is not random; it is constrained by knowledge of the prize’s location. This conditional structure makes the problem unintuitive and a classic example of how probability can defy common intuition, motivating the use of careful reasoning or simulation to determine the optimal strategy.

### Simulate the Monty Hall Problem

To determine whether it is advantageous to switch doors, we can simulate the game multiple times. The simulation follows a simple procedure:

1. Randomly place the car behind one of the doors.
2. Randomly select a door as the contestant’s initial choice.
3. Reveal a door with a goat, chosen from the remaining doors that do not hide the car.
4. Apply a strategy: either

    4.1 **Switch** to the other unopened door, or<br>
    4.2 **Stick** with the initial choice.

By repeating this process thousands of times, we can estimate the success rates for each strategy. Comparing the proportion of wins under the “stick” strategy versus the “switch” strategy allows us to evaluate which approach is more effective. Simulation thus provides an intuitive and practical method to study a probabilistic problem with conditional structure that can be difficult to reason about purely analytically.

````python
def monty_hall(n=10_000, switch=False):
    nb_wins = 0 
    doors = np.array(["GOAT", "CAR", "GOAT"])
    for _ in range(n):
        # Randomly place the car behind the doors by shuffling
        random.shuffle(doors)

        # Contestant randomly picks a car
        my_pick = random.choice([0, 1, 2])

        # Host opens a door with a goat (not the chosen door, not the car)
        remaining_doors = [i for i in range(3) if i != my_pick and doors[i] == "GOAT"]
        reveal_door = random.choice(remaining_doors)

        # Switch or stick
        if switch:
            my_pick = list(set([0, 1, 2]).difference([my_pick, reveal_door]))[0]

        # Check if the constestant wins
        if doors[my_pick] == "CAR":
            nb_wins += 1
    return nb_wins
````

````python
monty_hall(n=100_000, switch=False)
````

### Which Strategy is Better?

The results of a single simulation — such as 100,000 trials — do not, by themselves, prove which strategy is universally better. They merely report the observed outcomes: for example, in one run, we may have won $nw_1$ times by switching and $nw_2$ times by sticking with the initial choice, with $nw_1 > nw_2$ in that instance.
  
The natural questions then are: **Is switching always better?** How often can we expect the “switch” strategy to outperform “stick”? To answer this, we must consider the long-run probability that $nw_1 > nw_2$. In other words, over many repetitions of the simulation, what proportion of runs will result in switching yielding more wins than sticking?
  
This perspective emphasizes the difference between the outcome of a single random sample and the underlying probabilistic tendency of the generative process. While individual runs may vary due to chance, the probability that switching produces a better result can be quantified using repeated simulations or analytical reasoning. In the Monty Hall problem, theory and long-run simulations both indicate that switching is superior in most runs, reflecting the underlying structure of the problem rather than any single sample outcome.

````python
def compare_strategy(batches=1_000, trials_per_batch=10):
    """
    Estimate probability that switching is better than sticking.
    """
    nb_better_occurr = 0
    for i in range(batches):
        nb_wins_switch = monty_hall(trials_per_batch, switch=True)
        nb_wins_stick = monty_hall(trials_per_batch, switch=False)
        if nb_wins_switch >= nb_wins_stick:
            nb_better_occurr += 1 
    return(nb_better_occurr / batches)
````

## Working with Non-Uniform Probabilities

Simulating events with uniform probabilities is straightforward because each outcome in the sample space has an equal chance of occurring. For example, tossing a fair coin or rolling a fair die can be easily simulated using Python’s `random.choice()` function, where we pass a list of all possible outcomes.

However, many real-world situations involve non-uniform probabilities, where some outcomes are more likely than others. Examples include a biased coin, the popularity of products in a catalog, or customer choices in an online store. To simulate such scenarios, we need to account for the different probabilities of each outcome. 

### Working with Non-Uniform Probabilities -- An Example

Consider a practical scenario in which you are the CEO of a startup preparing for a funding round for your new software. You want to demonstrate to investors that users value some newly added functionality, and you decide that testing results from at least 30 users is the minimum number worth reporting.

Suppose the probability of any individual responding to your request to volunteer is 0.03. On average, this means that out of 100 people who receive your brochure by email, about three will volunteer. You don’t have a pre-existing marketing database, and sending emails costs money. Within your budget, you can afford a package that sends 1,200 emails. The key question becomes: what is the probability that at least 30 people will volunteer if you send these 1,200 emails?

To answer this, you can simulate the process using a simple probabilistic model. First, encode volunteers as `1` and non-volunteers as `0`. Next, create a population vector that reflects the proportions of participants (1s) and non-participants (0s) in the population. Finally, use `numpy.random.choice()` to randomly sample 1,200 individuals from this population multiple times and estimate the probability that at least 30 will participate. This approach allows you to account for the non-uniform probability of volunteering and assess the likelihood of meeting your target number of testers.

````python
# Example of generating a population vector with equal probability
population = [0] * 5  + [1] * 5
population
````

````python
random.choice(population)
````

````python
individuals = {1: "Volunteer", 0: "Non-volunteer"}
for i in range(4):
    val = random.choice(population)
    print(f"select {val}, which represents a {individuals[val]}")
````

````python
# 1. Encode those who volunteer as 1 and those who don't as 0, and
# 2. Generate a population vector that represents the ratio of users 
#    likely to volunteer. 
#    In 100 people, we have (voluteer probability = 0.03):

population = [0] * 97  + [1] * 3
````

````python
# The code below can be easily written using numpy.random.choice() method

def get_random_sample(population, nb_samples):
    nb_volunteers = 0 
    for _ in range(nb_samples):
        nb_volunteers += random.choice(population)
    return nb_volunteers
````

````python
get_random_sample(population, 1200)
````

````python
# Recall that the probability is the long-run proportion with which 
# a certain event will occur.
# 3. Repeat the experiment and count the number of times you have 
# at least 30 volunteers, i.e., satisfied with the outcome.

nb_satisfied = 0
nb_trials = 10_000
nb_volunteers_per_exp = []
for i in range(nb_trials):
    nb_volunteers = get_random_sample(population, 1200)
    nb_volunteers_per_exp.append(nb_volunteers)
````

````python
# 3. Probability that at least 30 will participate
nb_satisfied = sum(np.array(nb_volunteers_per_exp) > 30)
print(nb_satisfied/nb_trials)
````

````python
plt.figure(figsize=(14, 6))
plt.hist(nb_volunteers_per_exp, edgecolor='black', linewidth=1.2)
plt.show()
````

### Probability Models and Probability Distributions 

Many common probabilistic problems, such as the Monty Hall scenario or predicting the number of volunteers in an email campaign, can be simulated manually, but there are easier and more efficient ways to compute probabilities. Instead of manually generating a population vector and sampling from it, we can use parameterized probability functions provided by libraries like NumPy.

These functions, given the appropriate parameters, can directly compute the probability of any outcome in the sample space or generate random outcomes according to the defined probability model. Such functions correspond to **probability distributions**, which are mathematical models describing how probabilities are assigned to all possible outcomes of an experiment. Using probability distributions simplifies simulation, allows for scaling to larger problems, and provides a more general framework for reasoning about uncertainty.
