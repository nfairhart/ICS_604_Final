## Reproducing a Plot

1. Use `numpy`'s `random` module to generate a random list of size 1000 of the following four categories: "Physics", "CS", "Math" and "Engineering".
    * Use a seed of `42`.<br><br>
    
2. Store the data in an array called `students_affiliations`.
3. Use `matplotlib` or `pandas` to plot the data in `students_affiliations` to reproduce the following graph.

    * The plot uses `"#FF5733"` as color with an alpha (transparency) of `0.55`.

Your graph should look exactly like the one below.

<center><img src="ex_vis_figure.png" style="width:600px">

```python
## using pandas & np.random.choice

import pandas as pd
import numpy as np

disciplines = ['CS', 'Engineering', 'Math', 'Physics']

np.random.seed(42)
students_affiliations = pd.Series(np.random.choice(disciplines, 1000, replace=True)).value_counts()

ax = students_affiliations.plot.bar(color='#FF5733', alpha=0.55, width=0.8, rot=0)
ax.set_title("Distribution of Students in the Program", color='green')
ax.set_xlabel('Disciplines')
ax.set_ylabel('Counts');
```

```python
## using pandas & np.random.randint

import pandas as pd
import numpy as np

disciplines = ['CS', 'Engineering', 'Math', 'Physics']

np.random.seed(42)
students_affiliations = np.random.randint(0, 4, 1000)

sa = pd.Series(students_affiliations).value_counts()
sa.index = [disciplines[i] for i in sa.index]

ax = sa.plot.bar(color='#FF5733', alpha=0.55, width=0.8, rot=0)
ax.set_title("Distribution of Students in the Program", color='green')
ax.set_xlabel('Disciplines')
ax.set_ylabel('Counts');
```

```python
## using matplotlib & np.random.randint

import numpy as np
import matplotlib.pyplot as plt

disciplines = ['CS', 'Engineering', 'Math', 'Physics']

np.random.seed(42)
students_affiliations = np.random.randint(0, 4, 1000)

# count the occurence of unique values
_, counts = np.unique(students_affiliations, return_counts=True)
print(counts)

# get indices that would sort counts in decreasing order
idx = np.argsort(-counts)
disciplines = [disciplines[i] for i in idx]

plt.bar(disciplines, counts[idx], color='#FF5733', alpha=0.55)
plt.title("Distribution of Students in the Program", color='green')
plt.xlabel("Disciplines")
plt.ylabel("Counts")
plt.show()
```
