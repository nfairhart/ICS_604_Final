## The Normal Distribution

The **normal distribution**, also known as the **Gaussian distribution**, is a continuous probability distribution commonly used to model real-world phenomena. Many natural and human-made processes produce measurements that can be approximated by a normal distribution due to the central limit theorem.

Examples of such phenomena include:
* Heights of individuals within a population
* Measurement errors from instruments
* Duration of cab rides or other time-based processes

The normal distribution is especially important in statistics because many standard tests and methods assume that data are approximately normally distributed. This assumption underlies methods for estimating probabilities, confidence intervals, and hypothesis testing.

### Properties of the Normal Distribution

The normal distribution is a continuous probability distribution, often represented graphically as a smooth, bell-shaped curve. Unlike discrete distributions, continuous distributions can take any value within a range — either closed or open — making their probability space uncountably infinite.

<center><img src="https://www.dropbox.com/scl/fi/xdpm61o5k68muzqb0wzea/gaussian.png?rlkey=dnmoffwceokv49qzfyb7qp8ui&st=j9dq3p7v&dl=1" width="400"></center>

The bell shape is symmetric, with the peak (mean) located at the center. The shape and position of the curve are determined by two parameters:

* __Mean__ ($\mu$): Determines the center or location of the distribution along the x-axis.
* __Standard deviation__ ($\sigma$): Determines the spread of the distribution, controlling how wide or narrow the bell is. This parameter is sometimes referred to as the scale.

We denote that a random variable $X$ follows a normal distribution with mean $\mu$ and standard deviation $\sigma$ as:
  
$$
X \sim \mathcal{N} (\mu,\sigma)
$$

Just like with the binomial distribution, changing the parameters $\mu$ and $\sigma$ produces a different distribution, each telling a different story:

$$
X \sim (0, 1) \ne Y \sim (1, 1) \ne Z \sim (2, 2)
$$

````python
from scipy.stats import norm

plt.figure(figsize=(10, 4))

x = np.arange(-3.5, 3.5, 0.05)
p_x = [norm.pdf(val, 0, 1) for val in x]
plt.plot(x, p_x, label="$\\mu = 0$ and $\\sigma = 1$")

x = np.arange(-13, 17, 0.05)
p_x = [norm.pdf(val, 2, 5) for val in x]
plt.plot(x, p_x, label="$\\mu = 2$ and $\\sigma = 5$", color='r')

plt.legend();
````

### The Probability Density Function (PDF) for a Normal Distribution

For a random variable $X$ distributed according to a normal distribution with mean $\mu$ and standard deviation $\sigma$, $X \sim \mathcal{N} (\mu,\sigma)$, the **probability density function (pdf)** is given by::

$$ 
    p(x) = \frac{1}{\sqrt{2\pi}\sigma} e^{\frac{(x-\mu)^2}{2\sigma^2}}
$$

Key points to understand:

* Unlike discrete random variables, the sample space of a continuous random variable is uncountably infinite.
* This means the probability of observing any single, exact value is 0.
* Instead of a probability mass function (`pmf`), continuous random variables use a probability density function (`pdf`).
  * The `pdf` describes how probability is distributed across intervals of values.
  * Probabilities are obtained by integrating the PDF over a range of interest:

$$
\Pr(a \le X \le b) = \int_a^b p(x) \, dx
$$

This shift from `pmf` to `pdf` is what allows continuous distributions, like the normal, to describe probabilities over continuous ranges rather than discrete points.

````python
# We can manually implement the PDF

def computePDF_normal (x, mu, sigma):
    return 1.0/(sigma * np.sqrt(2 * np.pi)) * np.exp(-(x - mu)**2.0 / (2 * sigma**2))
````

### `pdf` versus `pmf`

A probability density function (`pdf`) plays a role similar to a probability mass function (`pmf`), but for continuous random variables. Instead of assigning a probability to a single outcome, a pdf describes how densely probability is distributed around a particular value. Probabilities are obtained from the areas under the curve over intervals of values. 

For example, for a random variable $X \sim \mathcal{N}(0, 1)$, the density near $x=0$ is much higher than the density near $x=−2.5$, indicating that values close to 0 are far more likely to occur than values far from 0.

Because $X$ is continuous, the pdf itself is a continuous function. The probability of observing any exact single value is zero; instead, probabilities are obtained by integrating the pdf over an interval. The total area under the pdf curve over the entire sample space is equal to 1, representing total probability.

### Data in the Distribution: 1$\sigma$

In a normal distribution, the region of highest probability density lies close to the mean $\mu$. As a result, when sampling from a normal distribution, most observations tend to cluster near the mean. Specifically, approximately **68%** of the data fall within one standard deviation of the mean, that is, in the interval $[\mu - \sigma, \mu + \sigma]$.

### Data in the Distribution: 2$\sigma$

As we move farther away from the mean, the probability density decreases. Nevertheless, a substantial portion of the data still lies within two standard deviations of the mean. Approximately **95%** of observations fall within the interval 
$[\mu - 2\sigma, \mu + 2\sigma]$, capturing most of the variability in the distribution while excluding more extreme values.

### Data in the Distribution: 3$\sigma$

Values far from the mean occur with much lower probability in a normal distribution. Nearly all observations are contained within three standard deviations of the mean. In fact, approximately **99.7%** of the data lie in the interval 
$[\mu - 3\sigma, \mu + 3\sigma]$. This empirical rule highlights that observations outside this range are rare and are often considered extreme or outliers in practice.

### Computing the `pdf` in Python

For continuous random variables, probability density functions can be evaluated directly using the `scipy.stats` module. SciPy provides built-in implementations of many common probability distributions, including the normal (Gaussian) distribution.

```python
import scipy.stats
norm_0_1 = scipy.stats.norm(0, 1)
print(norm_0_1.pdf(2))
```

In this example, `scipy.stats.norm(0, 1)` creates a frozen normal distribution with mean 0 and standard deviation 1. The `pdf` method then evaluates the probability density at a specific point, in this case $x=2$.

The `pdf` method can also be applied to a collection of values at once:

```python
print(norm_0_1.pdf([-2, -1, 0, 1, 2]))
```

This returns the density at each specified point along the $x$-axis.

Alternatively, the same computation can be performed without creating a frozen distribution by calling the function directly and specifying the parameters explicitly:

```python
scipy.stats.norm.pdf(0, loc=0, scale=1)
```
Because the normal distribution is symmetric and centered at the mean, the density is highest at $x=0$ and decreases as we move away from the mean. Consequently, `pdf(0)` is the largest among the values shown. Moreover, due to symmetry about the mean, `pdf(-2)` and `pdf(2)` are equal.

````python
norm_0_1 = scipy.stats.norm(0, 1)
print(norm_0_1.pdf(2))
````

````python
print(scipy.stats.norm.pdf(2, 0, 1))
````

````python
print(norm_0_1.pdf([-2, -1, 0, 1, 2]))
````

````python
print(scipy.stats.norm.pdf([-2, -1, 0, 1, 2], 0, 1))
````

````python
# return 3 random variates from Normal with mean 0 and std = 1

print(norm_0_1.rvs(3))
print(scipy.stats.norm.rvs(loc=0, scale=1, size=3))
````

### The Sum of PDF $\neq$ 1

```python 
>>> print(sum(norm_0_1.pdf([-2, -1, -0.5, 0, 0.5, 1, 2])))

1.6949963159946946
```

For continuous random variables, the probability density function (pdf) does not assign probabilities to individual outcomes. Instead, it describes how *dense* the probability mass is around a given value. As a result, the numerical values returned by a pdf are not constrained to sum to one when evaluated at a finite set of points.

In the example above, we compute the pdf of a standard normal distribution at a small collection of values and then add those density values together. This sum exceeds 1 because each pdf value represents the height of the density curve at a single point, not the probability of that outcome occurring. For continuous distributions, the probability of observing any exact value is zero.

Probabilities are obtained by **integrating the pdf over an interval**, which corresponds to computing the area under the curve over that range. Only the total area under the entire pdf curve — integrated over the full sample space — equals 1. Summing pdf values without accounting for an interval width does not represent a probability and therefore has no reason to equal (or be less than) 1.

````python
print(sum(norm_0_1.pdf([-2, -1, -0.5, 0, 0.5, 1, 2])))
````

````python
norm_0_1.pdf([-2, -1, -0.5, 0, 0.5, 1, 2])
````

````python
print(scipy.stats.norm.pdf(1, loc=1, scale=0.1))
````

### Plotting the `pdf`

When working with continuous probability distributions, it is not possible to compute the probability density function (pdf) at every possible value, since the sample space is uncountably infinite. However, for visualization purposes, this is not necessary. Instead, we can evaluate the pdf at a sufficiently large and finely spaced set of points so that the resulting plot appears smooth and continuous.

A common approach is to generate a grid of evenly spaced values over a range of interest and then compute the pdf at each of those points. In Python, this is often done using `np.arange`, which creates an array of values starting at `start`, ending before `stop`, and incremented by a fixed `step` size:

```python
np.arange(start, stop, step)
```

As with Python’s built-in `range` function, the `stop` value is not included. By choosing a small step size (for example, 0.01 or 0.001), we obtain enough points to accurately approximate the continuous curve of the pdf when plotted.

````python
np.arange(-2, 2, 0.5)
````

````python
plt.figure(figsize=(8, 4))

norm_0_1 = scipy.stats.norm(0, 1)
x_axis = np.arange(-2, 2.5, 0.5)
y_axis = norm_0_1.pdf(x_axis)

_ = plt.plot(x_axis, y_axis)
````

````python
plt.figure(figsize=(8, 4))

norm_0_1 = scipy.stats.norm(0, 1)
x_axis = np.arange(-4, 4.05, 0.05)
y_axis = norm_0_1.pdf(x_axis)

_ = plt.plot(x_axis, y_axis)
````

## Illustrating Density

To illustrate how probability density relates to the values we observe when sampling from a continuous distribution, we can draw random samples from a normal distribution. Consider sampling 1,000 values from a normal distribution with mean $\mu=0$ and standard deviation $\sigma=2$, written as $\mathcal{N}(0, 2)$. Because regions near the mean have higher probability density, we expect more sampled values to fall near zero than far away from it.

We can visualize this behavior by plotting the sampled values as a histogram. The histogram groups the data into a fixed number of bins — in this case, 20 — and displays the count of observations in each bin on the $y$-axis. Bins near the center of the distribution will typically have higher counts, reflecting the higher density of the normal distribution in that region, while bins farther from the mean will have fewer observations. This provides an intuitive, empirical view of how density translates into observed frequencies in random samples.

````python
# Ex. sampling 1000 values from Gaussian with mean 0 and standard deviation 2

np.random.seed(225)
gaussianSamples = np.random.normal(0, 2, 1000)
gaussianSamples[0:10]
````

````python
len(gaussianSamples)
````

````python
# plot the counts rather than the density

plt.figure(figsize=(10, 4))
counts, _, _ = plt.hist(gaussianSamples, bins=20, edgecolor="k", linewidth=2, alpha=0.4)
````

````python
print("Total count =", np.sum(counts))
````

````python
# plots the density rather than the counts

plt.figure(figsize=(10, 4))
density, bins, _ = plt.hist(gaussianSamples, bins=20, edgecolor="k", linewidth=2, alpha=0.4, density=True)
````

````python
sum_density = np.sum(density)
print("Sum of densities =", sum_density)
print("Area under the curve =", sum_density * np.diff(bins)[0])
````

### Comparing the Density Curve and the Histogram

When we overlay the probability density function (pdf) on top of a histogram of sampled data, we are essentially comparing the theoretical distribution to an empirical approximation. The histogram shows how many values fell into each bin, while the pdf shows the idealized density across the continuous range of values.

The histogram should not match the pdf exactly, especially if the number of samples is limited or the bin width is coarse. With more samples and finer bins, the histogram will more closely approximate the shape of the pdf. Minor differences are expected due to the randomness inherent in the sampling process — histograms are empirical, pdfs are theoretical.

````python
plt.figure(figsize=(10, 4))

normal_0_2 = scipy.stats.norm(0, 2)
x_axis = np.linspace(-8, 8, 100)
y_axis = normal_0_2.pdf(x_axis)

plt.plot(x_axis, y_axis, color="k")

# Note that the param density=True is necessary here
_ = plt.hist(gaussianSamples, bins=20, edgecolor="k", color='#80d4ff', linewidth=2, alpha=0.2, density=True)
````

### The Sum of `pdf`s

In a continuous distribution, the probability of observing *any exact single value* is 0 because the sample space is uncountably infinite. That’s why, unlike a discrete probability mass function (pmf), summing the values of a probability density function (pdf) *does not give 1*.

The pdf instead represents **density**, which is probability per unit of the x-axis. If we want the actual probability over a range, we multiply the density by the width of the interval:

$$
\text{mass (probability)} = \text{density} \times \text{interval width}
$$

This is equivalent to computing the **area under the curve** over the interval (for example, $x-0.01 < x <  x+0.01$). For the entire distribution, the total area under the pdf curve sums to 1, even though the sum of the densities at discrete points can be greater than 1.

````python
norm_cdf_sum = 0

for x in np.arange(-8, 8, 0.0001):
    norm_cdf_sum += computePDF_normal(x, 0, 2) * 0.0001
    
print("The sum of the probability densities computed above is:", norm_cdf_sum)

# Use of the cumulative distribution function (cdf) 
print("The sum of the probability densities computed using cdf:", norm.cdf(8, 0, 2))
````
