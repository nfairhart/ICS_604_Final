## Exploring Your Data

Data analysis typically begins with exploring the dataset to gain an understanding of its structure, quality, and characteristics. ### Pandas Functionality

Pandas DataFrames offer a variety of features for efficiently working with tabular data. These include **attributes**, which describe constants or properties of the data, such as the number of rows or the data type of a specific column. DataFrames also provide **functions that act on the data**. Some of these functions are non-transformative, returning information without modifying the DataFrame — for example, finding the maximum value in a column. Others are transformative, meaning they return a modified version of the DataFrame without changing it in place, such as removing a column, sorting by a column, or filtering rows based on a condition.

In addition, Pandas supports applying **custom functions** to the data, allowing you to perform user-defined operations. For instance, you can normalize the values in a column by subtracting the column mean and dividing by its standard deviation, making the data ready for further analysis or modeling.

### DataFrame Axes

Computing statistics is a key part of exploratory data analysis, as it provides valuable insights into the dataset — for example, identifying the most prescribed medications, the doctors or specialties generating the most prescriptions, or the total opiate prescriptions per state. While we often perform computations column-wise, there are situations where row-wise operations are needed. Instead of creating separate functions for each direction (e.g., `compute_avg_row()` vs. `compute_avg_col()`), Pandas uses the concept of an axis. Many methods, such as `sum()`, `min()`, and `max()`, can operate along either rows or columns by specifying the axis parameter. By default, `axis=0`, which applies the operation column-wise (along rows). This approach allows a single method to handle both row-wise and column-wise computations efficiently, though the concept may take some practice to fully understand.

### Descriptive Statistics

Exploring functions on a simple, manually created DataFrame is a great way to test and understand Pandas functionality. Making a small test dataset is especially useful when you are unsure how a method works after reading the documentation. For example, consider the following dataset:

````python
df = pd.DataFrame({'A':[0, 1, 2], 'B':[5, 6, 7]})
df
````

By default, the `mean()` method computes column-wise averages:

````python
print('Column-wise mean:')
df.mean()
````

Notice that the result is a Series. To compute row-wise averages instead, you can set the `axis` parameter to 1:

````python
print('Row-wise mean:')
print(df.mean(axis=1))
````

Descriptive statistics can also be applied directly to a Series. For instance, if you are interested only in column `A`, you can extract it and compute its mean:

````python
print('Series mean:')
print(df['A'].mean())
````

Working with small test datasets like this helps clarify how methods behave and how the `axis` parameter affects computations.

### Missing Values

In real-world datasets, missing values can appear in many different forms, such as blank spaces, empty strings, or predefined placeholders like `MISSING` or `NOVAL`. These values can significantly skew results or even break an analysis. For example, the presence of a placeholder like `NOVAL` in an otherwise numeric column can cause Pandas to treat the entire column as an object type, preventing numerical computations. As a result, custom handling of missing values is often necessary, which typically involves converting these placeholders into Pandas’ standard missing value representation, `NaN`. In Pandas, `NaN` stands for “Not a Number” and represents the absence of a value. It denotes undefined or unrepresentable data and is distinct from valid values such as `0`, an empty string, or the boolean `False`.


````python
np.nan, type(np.nan)
````

### Counting Missing Values

Although this DataFrame contains no actual data values, Pandas still reports a shape of `(2, 2)` and a size of `4`, reflecting the number of rows, columns, and total cells—not the number of valid entries.

````python
na_df.shape
````

````python
na_df.size
````

To assess how much usable data is present, it is often more informative to use the built-in `count()` method. Unlike `shape` and `size`, `count()` automatically excludes missing values and returns the number of non-missing entries per column. This makes it an easy and effective way to evaluate the extent of missing data in a DataFrame before proceeding with further analysis.

````python
na_df.count()
````

### Sorting

Sorting data is an important step during data exploration. It can help detect anomalies, understand the scales and ranges of values (such as minimum and maximum transactions), and identify important categories.

In Pandas, sorting can be performed in two main ways. You can sort a DataFrame by its row index using the `sort_index()` method. Since a DataFrame has only one index, you do not need to specify which index to sort — the method automatically sorts based on the row labels. Alternatively, you can sort by one or more column values using `sort_values()`. Because a DataFrame may have many columns, you must explicitly specify the column(s) to sort on.

By default, sorting is done in ascending order. You can modify this behavior by setting the `ascending` parameter to `False` for descending order

````python
spending_df.sort_index()
````

````python
spending_df.sort_values('spending', ascending=False)
````

### Method Chaining

Method chaining allows multiple operations to be applied sequentially to the same object within a single line of code. Rather than storing intermediate results in separate variables, each method called on the output of the previous one. The general syntax for method chaining is:

```python
object.method_1().method_2()...
```

This style often results in more concise and readable code, especially for data-processing workflows. We will demonstrate method chaining using the `sample()` method, which — although not yet formally introduced — is intuitive and can be used to randomly select a specified number of rows from the data.

````python
spending_sorted_df = spending_df.sort_values(by="spending", ascending=False)
spending_sorted_top_10_df = spending_sorted_df.head(10)

spending_sorted_top_10_df
````

````python
spending_sorted_top_10_df.sample(2)
````

````python
spending_df.sort_values(by="spending", ascending=False).head(10).sample(2)
````

Chained method calls are often written across multiple lines to improve readability, especially when several operations are applied in sequence. However, Python requires a specific syntax when breaking chained methods across lines. Unlike languages such as JavaScript, where statements typically end with a semicolon, Python expects each statement to end at the end of the line unless explicitly told otherwise. As a result, placing a line break immediately after a dot (`.`) without additional syntax will raise a syntax error. The example shown illustrates this issue, as Python interprets the line break as the end of the statement and fails to parse the chained methods correctly.

````python
spending_df.
            sort_values(by="spending", ascending=False).
            head(10).
            sample(2)
````

To resolve this issue, Python allows method chaining to be split across multiple lines by wrapping the entire expression in parentheses. This tells Python that the statement continues onto the next line, allowing each method call to be written on its own line and improving readability.

````python
(spending_df.
            sort_values(by="spending", ascending=False).
            head(10).
            sample(2))
````
