## Comparing Proportions

Another application of hypothesis testing involves comparing an observed proportion to a known population proportion. According to data published on Statista, the national ratio of branded to unbranded generic medications is 1:5. To explore ways of controlling healthcare spending, an NGO tested a set of incentives aimed at increasing the use of unbranded generic medications in a specific geographical area over the course of one year. At the end of the trial period, a survey of medications sold in that area found that 3,766 out of 4,434 drugs sold were generics.

The hypothesis testing question here is whether this result differs significantly from the national proportion. Importantly, we are not yet asking whether the incentive program was effective or successful — only whether the observed proportion in the target area is statistically different from the national benchmark. This distinction matters, as a statistically significant difference could reflect the program's impact, but could also be driven by other local factors.

### Testing Proportion

A national branded-to-generic ratio of 1:5 means that 5 out of every 6 drugs sold are generic, corresponding to approximately 83% of all medications. In our survey, 3,766 out of 4,434 drugs were generic, yielding a sample proportion of roughly 84.9%.

The key question is whether this 1.9 percentage point difference from the national proportion can be explained by random sampling variability alone, or whether it might reflect a genuine effect of the NGO's incentive strategy. To answer this, we examine the distance between the observed sample proportion of 84.9% and the national benchmark of 83%. The larger this distance, the harder it becomes to attribute the difference to chance — and the stronger the evidence that the strategy is producing a real shift in medication purchasing behavior.

### Deciding on the Test Statistic

To formally test whether the observed difference is meaningful, we use the absolute distance between the sample proportion of generic drugs and the national benchmark of 83% as our test statistic:

$$
∣(\text{sample \% of generic drugs})  − 83∣
$$

Under the **null hypothesis**, we assume there is no real difference between the proportion of generic drugs sold in the surveyed area and the national proportion — meaning any observed difference is simply the result of sampling variability. The **alternative hypothesis**, on the other hand, posits that the observed difference is too large to be plausibly explained by chance alone.

### Simulating Data to Build the Sampling Distribution of the Test Statistic

To evaluate our null hypothesis, we need to understand what values of the test statistic are plausible under it — in other words, how large a difference between a sample proportion and the national mean we would expect to see simply due to random sampling. To do this, we simulate samples drawn from outside the testing area, where the incentive program was not applied, and compute the test statistic for each sample. This builds an empirical sampling distribution of the test statistic under the null hypothesis, against which we can compare our observed result.

- G: Generic, B: Brand Name

- Beyond collecting another physical sample from the surveyed area, what other strategies can we use to explore the range of differences between 83% and samples drawn from the same distribution? In other words, how can we generate random samples that mimic those produced under the null hypothesis — without the need to physically collect new data?

```python
# First example
model_proportions = [0.17, 0.83]

data = np.random.choice(["B","G"], p=model_proportions, size=4434)
data
```

```python
print(sum(data == "G") / 4434 * 100)
```

```python
# Or using the binomial distribution
np.random.binomial(4434, 0.83) / 4434 * 100
```

```python
sample_diffs_null = []
for _ in range(5000):
    sample_proportion = sum(np.random.choice([0, 1], p=model_proportions, size=4434)) / 4434 * 100
    sample_diff = np.abs(sample_proportion - 83)
    sample_diffs_null.append(sample_diff)

plt.figure(figsize=(8, 4))
plt.hist(sample_diffs_null, bins=15, edgecolor='k', linewidth=1.4);
```

```python
plt.figure(figsize=(8, 4))

plt.hist(sample_diffs_null, bins=15, edgecolor='k', linewidth=1.4)
plt.scatter(1.9, 0, color='r', s=80)
plt.ylim(-30, 1050);
```

```python
# computimg the p-value
p_value = sum(np.array(sample_diffs_null) >= 1.9) / len(sample_diffs_null)
p_value
```

## Comparing Multiple Categories

When working with data that fall into more than two categories, the same hypothesis-testing ideas used in the previous example can still be applied. Instead of comparing a single proportion, we now compare the distribution of observations across several categories to an expected distribution. The null hypothesis states that the observed counts in each category follow the expected proportions, while the alternative hypothesis suggests that at least one category differs from what we would expect.

The overall approach remains largely the same as before. We begin by specifying the expected proportions for each category based on theory, prior data, or national statistics. Then we collect our sample data and compute how many observations fall into each category. Finally, we evaluate whether the differences between the observed counts and the expected counts are large enough to suggest that the sample does not come from the assumed distribution.

The main difference in this situation is the need for a **new test statistic** that can measure discrepancies across multiple categories at once. Rather than comparing just two proportions or two means, the statistic must summarize how far the entire observed distribution deviates from the expected distribution. This test statistic allows us to quantify the overall difference and determine, through a reference or simulated null distribution, whether the observed pattern is likely due to random variation or indicates a meaningful departure from the expected model.

### Example: A New Method for Estimating Fish Diversity?

In this example, we evaluate a new machine learning–based method for estimating fish diversity in a pisciculture pond. The proposed approach uses environmental DNA (eDNA) collected from the water and analyzed with an automated model to estimate the relative proportions of different fish species.

<img src="https://www.dropbox.com/scl/fi/kiuuqt78d7gvb4cxtrkgp/eDNA_fish.png?rlkey=hk7dgnwsjvlhx4400rpqc6htr&st=yctauq2y&dl=1"><br>

The method produces estimated ratios for several categories of fish present in the pond: Tilapia, Blenny, Angelfish, Salmon, and a category labeled “Other.” Because this method is designed to be faster and more efficient than traditional approaches, it is important to determine whether its estimates are consistent with the current standard method.

To obtain a reliable baseline, fish were manually counted using the traditional approach. In this process, 500 fish were collected and individually categorized by species. Although this method is considered the state of the art, it is labor-intensive and time-consuming. The resulting counts provide the observed distribution of species in the pond and serve as the reference against which the new method will be compared.

Next, the eDNA-based system was used to estimate the proportions of each fish category in the same pond. The results from this automated method are summarized in the graph shown. By comparing the species ratios obtained from the manual counts to those estimated by the eDNA model, we can evaluate whether the new approach produces a distribution of species that is statistically consistent with the established method. If the differences between the two distributions are small and within the range expected from sampling variability, the new method may be considered a reliable and much more efficient alternative for estimating fish diversity.

```python
fish_proportions = pd.DataFrame({
    'kind': ['Tilapia', 'Blenny', 'Angelfish', 'Salmon', 'Other'], 
    'sampled': [0.20, 0.08, 0.12, 0.54, 0.06], 'eDNA': [0.26, 0.08, 0.08, 0.54, 0.04]})
fish_proportions
```

```python
fish_proportions.plot.barh(x="kind", figsize=(10, 5));
```

### Questions

The distributions shown in the graph appear to differ slightly between the manually collected counts and the estimates produced by the eDNA-based method. However, small differences between two distributions can arise simply due to random variation. To determine whether the observed differences are meaningful, we must frame the problem using a formal hypothesis-testing approach.

Before carrying out the analysis, it is helpful to identify the key questions that will guide the process. These questions structure how we evaluate whether the new eDNA-based method produces results consistent with the traditional manual counting approach. Specifically, we need to determine what hypotheses we should test, how to quantify the difference between the two distributions, and how to model what the data would look like if the two methods were actually consistent.

To guide the analysis, we will address the following questions:

- What are the null and alternative hypotheses?
- What is a useful test statistic for this problem?
- How can we generate data following the null hypothesis?

### 1. Null and Alternative Hypotheses

To begin the analysis, we first define the hypotheses that will guide our statistical test. These hypotheses describe the competing explanations for the differences observed between the manual counting method and the eDNA-based estimates.

The **null hypothesis** states that the observed differences between the two methods are due to **sampling variability**. In other words, the eDNA estimates and the manual counts represent the same underlying distribution of fish species, and any discrepancies between the two arise purely by chance. Under this assumption, if we were to repeat the sampling process multiple times, we might observe similar differences — or even larger ones — simply because of random variation in the sample.

The **alternative hypothesis** states that the observed difference between the two methods is **too large to be explained by random chance alone**. If this were true, it would suggest that the eDNA-based method produces a distribution of species that differs systematically from the one obtained through manual counting. In that case, repeating the sampling process would rarely produce a difference as large as the one observed.

It is important to note that our goal here is not to determine whether the new method overestimates or underestimates any particular fish species. Instead, the focus is on whether the overall distribution of species estimated by the new method is statistically consistent with the distribution obtained through manual counting of a large sample.

### 2. The Test Statistic

A good test statistic for this problem is one that allows us to quantify the difference between the two distributions. Since we are comparing the proportions of several fish categories produced by two different methods, the statistic should measure how far apart the two sets of proportions are overall.

#### The Distance Between Two Distributions

One way to measure this difference is by computing the **total variation distance (TVD)** between the two distributions. Conceptually, this measure is similar to the Euclidean distance between two vectors, where each vector represents the proportions of the fish categories produced by a given method. The TVD summarizes how much the probabilities assigned by the two methods differ across all categories.

```python
fish_proportions["difference"] = fish_proportions["sampled"] - fish_proportions["eDNA"]
fish_proportions
```

#### The Total Variation Distance


When we calculate the difference between the proportions from the two methods for each fish category, the resulting values will sum to zero. This happens because both distributions represent proportions that add up to 1. For example, the positive differences might add up to 0.06, while the negative differences sum to −0.06, perfectly canceling each other out. This reflects the fact that both the basin (sampled) and collected (eDNA) distributions sum to 1, so their total difference must equal to 0 (= 1 − 1).

To prevent positive and negative values from canceling each other, we instead sum the absolute values of the differences and divide the result by two. This produces the total variation distance, defined as:

$$
tvd = \frac{1}{2}\sum_{x \in \Omega}|P(x) - Q(x)|,
$$

where $P$ and $Q$ represent the two probability distributions over the set of categories $\Omega$. In our case, these correspond to the proportions of fish species estimated by the manual sampling method and the eDNA-based method. While other distance measures could also be used, the total variation distance is both intuitive and computationally simple, making it a convenient choice for comparing categorical distributions.

```python
observed_tvd = 1/2 * sum(np.abs(fish_proportions['sampled'] - fish_proportions['eDNA']))
observed_tvd
```

### 3. Generating Data Under the Null Hypothesis

To evaluate whether the observed difference between the two methods is unusual, we need a way to simulate data **assuming the null hypothesis** is true. Under the null hypothesis, the distribution of fish species estimated by the eDNA method correctly reflects the true proportions in the pond. Using these proportions, we can generate synthetic datasets that represent what we would expect to observe if the manual sampling were drawn from that same distribution.

To understand the idea, consider a simplified case with only two categories: `Tilapia` and `Other`. Suppose the proportion of Tilapia is $p(Tilapia)=0.2$, which means the proportion of Other fish is $p(Other)=1−0.2=0.8$. If we want to simulate 500 observations, where each observation can be either Tilapia or Other, we can generate these outcomes using the binomial distribution, which models repeated trials with two possible outcomes.

```python
sample_1 = np.random.choice(['Tilapia', 'Other'], p=[0.20, 0.80], size=500)
print(sample_1[0:10])
print(sum(sample_1 == "Tilapia"))
```

```python
print(np.random.binomial(500, 0.20))
print(np.random.binomial(500, 0.20))
```

<br>

#### The Multinomial Distribution

In our actual problem, however, there are **more than two categories** (Tilapia, Blenny, Angelfish, Salmon, and Other). To simulate this type of data, we use the **multinomial distribution**, which is a generalization of the binomial distribution to multiple possible outcomes.

The multinomial model assumes that:

- Each trial is independent of the others.
- The probabilities of the outcomes remain constant across trials.
- The experiment consists of $n$ repeated trials.
- Each trial results in one of several discrete categories.

```python
# Throw a dice 120 times
np.random.multinomial(120, pvals=[1/6, 1/6, 1/6, 1/6, 1/6, 1/6] ) 
```

<br>Using the multinomial distribution, we can simulate many samples of fish counts with the same total sample size as our manual dataset. These simulated datasets represent what we would expect to observe if the null hypothesis were true, and they can be used to construct the **null distribution for our test statistic**.

```python
# This is equivalent to the multinomial returning the outcomes of each trial
np.random.choice(['Tilapia', 'Blenny', 'Angelfish', 'Salmon', 'Other'], 
                 p=[0.20, 0.08, 0.12, 0.54, 0.06], size=500)
```

### 4. Simulating the Experiment

Once we have a method for generating data under the null hypothesis, we can use it to **simulate the sampling distribution of the total variation distance (`tvd`)**. This allows us to understand how much variation in tvd we would expect if the eDNA-based method were perfectly consistent with the true species proportions.

The simulation proceeds as follows:

1. **Generate a random sample:** Using the multinomial distribution and the observed proportions from the manual method, simulate a new dataset with the same total sample size as the manual counts. Each trial represents the selection of a single fish, and the outcome is one of the species categories.

2. **Compute the `tvd`:** For this simulated dataset, calculate the total variation distance between the simulated proportions and the original proportions. This quantifies the discrepancy that arises purely due to random sampling.

By repeating these two steps many times, we generate a **null distribution of `tvd` values**. This distribution represents the range of differences we would expect if the null hypothesis were true. Comparing the observed `tvd` from the actual data to this null distribution allows us to assess how unusual the observed discrepancy is and provides the basis for a p-value to test whether the methods differ significantly.

```python
fish_proportions["sampled"]
```

```python
n = 500
samples_tvd = []

for i in range(5000):
    sample_proportions = np.random.multinomial(n, pvals=fish_proportions["sampled"]) / n
    sample_tvd = 1/2 * np.sum(np.abs(fish_proportions['sampled'] - sample_proportions))
    samples_tvd.append(sample_tvd)
```

```python
plt.figure(figsize=(10, 5))
plt.hist(samples_tvd, bins=15, edgecolor='k', linewidth=1.4, alpha=0.7);
```

```python
plt.figure(figsize=(10, 5))

counts = plt.hist(samples_tvd, bins=15, edgecolor='k', linewidth=1.4, alpha=0.7)
plt.scatter(observed_tvd, 2, color='r', s=100)
plt.ylim(-30, counts[0].max() + 50);
```

```python
p_value = np.sum([np.array(samples_tvd) >= observed_tvd]) / len(samples_tvd)
print(p_value)
```

### Notes About the Outcome

Using a **p-value threshold (significance level) of 0.05**, the results suggest that the distribution of fish species estimated by the two methods is statistically different. In the simulation, the observed total variation distance (`tvd`) of 0.06 falls far out in the tail of the null distribution histogram. This indicates that such a large discrepancy would rarely occur if the two methods were producing results from the same underlying distribution.

Because the observed statistic lies in this extreme region of the null distribution, we reject the null hypothesis. This suggests that the differences between the manual counts and the eDNA-based estimates are **too large to be explained by random sampling variability alone**.

However, it is important to note that this result does not explain why the distributions are different or what the difference necessarily implies about the methods. The statistical test only indicates that the two distributions are unlikely to be the same. Further investigation would be required to determine whether the discrepancy is due to limitations in the eDNA method, sampling issues, environmental factors, or other sources of bias.
