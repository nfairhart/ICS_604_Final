## Hypothesis Testing

In statistics, it is important to distinguish between statistics and parameters. A **statistic** is a numerical measure calculated directly from a sample of data. It summarizes or describes a characteristic of the sample and is often used to estimate or make inferences about the corresponding population characteristic. In contrast, a **parameter** is a numerical value that describes some aspect of the entire population. It is a fixed value, often unknown, that could be determined if data from the complete population were available.

A **hypothesis** is an assumption or statement about a population parameter, such as the population mean or proportion. The goal of a **hypothesis test** is to evaluate evidence from sample data in order to decide between two competing hypotheses — typically a null hypothesis and an alternative hypothesis. By systematically analyzing the data, hypothesis testing provides a formal framework for determining whether observed differences or effects are statistically significant.

Hypothesis testing methods can be categorized as **parametric** or **non-parametric**. Parametric methods make assumptions about the underlying population parameters and distributions, such as assuming normality. In contrast, non-parametric methods do not rely on assumptions about the parameters or the shape of the underlying distributions, making them more flexible when the population structure is unknown or does not meet parametric assumptions.

### Hypothesis Testing Questions

In hypothesis testing, we are generally concerned with two main aspects. The first is the **magnitude of the difference between hypotheses**, often referred to as the **effect size**. This measures how large or meaningful the observed difference is, beyond simply determining whether it exists. For instance, economists have studied the effect of a universal basic income on happiness. While the results may be statistically significant, the effect size answers the practical question: *how much happier are people actually?* If the observed increase in happiness is only 1%, policymakers may question whether spending $3 trillion — an estimated annual cost — is justified.

The second aspect is the **probability associated with the measured difference**. This involves understanding the variability of the observed difference and assessing how likely it is that the difference could have arisen purely by chance. This is captured by the **p-value**, which represents the probability of observing a difference at least as extreme as the one measured, assuming the null hypothesis is true. The p-value provides a quantitative measure of the statistical evidence against the null hypothesis.

### Data in Hypothesis Testing

Data used in hypothesis testing can come from either one or two samples. In the case of a single sample, the goal is often to compare the sample to a population with a known parameter, such as the mean. For example, if a new math curriculum is implemented with a randomly selected group of students, their test scores can be compared to the national average. The hypothesis test then evaluates whether the curriculum leads to a significant improvement in math exam scores compared to the broader population.

Alternatively, hypothesis testing can involve two samples, where the objective is to determine whether there is a difference between them. For instance, one group of students may receive the math curriculum in the morning, while another group receives it in the afternoon. The test then examines whether the time of day affects learning outcomes, allowing us to assess whether performance differs significantly between the two groups.

### Terminology

In hypothesis testing, we typically evaluate **two competing hypotheses**. The first is the **null hypothesis** ($H_0$), which assumes that there is no change or effect in the system and that any observed differences are simply due to natural variability in the population. The second is the **alternative hypothesis** ($H_a$), which assumes that the differences we observe are unlikely to have occurred by chance alone. By "unlikely," we mean that such outcomes would be rare if the null hypothesis were true.

The alternative hypothesis can take one of two forms. A **single-tailed hypothesis** specifies that the effect or difference occurs in a particular direction. For example, a new teaching method might improve the average math score compared to the national average, expressed as $\mu_{new\_method} \gt \mu_{national\_average}$. Similarly, a new medication might reduce cholesterol levels compared to a control group, written as $\mu_{cholesterol\_treatment} \lt \mu_{cholesterol\_control}$.

A **two-tailed hypothesis**, on the other hand, considers differences in either direction. Here, the values are expected to differ from a given reference but without specifying the direction. For instance, if we want to test whether the rate of opioid prescriptions in Hawaii differs from the national average, the alternative hypothesis would allow for either $\mu_{opioids\_Hawaii} \lt \mu_{opioids\_nationally}$ or $\mu_{opioids\_Hawaii} \gt \mu_{opioids\_nationally}$.

### About the Hypothesis Tests

In hypothesis testing, the null hypothesis ($H_0$) and the alternative hypothesis ($H_a$) are **mutually exclusive**. For example, either there is a difference or there is no difference, but it would not make sense to define the hypotheses as “the difference is greater than 0” versus “the difference is greater than 10.” Each hypothesis represents a distinct scenario about the population.

It is important to understand that a null hypothesis can never be proven true. Failing to find evidence for the alternative hypothesis does not confirm the null; it only makes the null hypothesis plausible given the data. In such cases, we say that we **fail to reject the null hypothesis** or that the study did not provide sufficient evidence to reject it.

Finally, we describe results as **statistically significant** when the observed outcome is unlikely to have occurred by chance alone. Statistical significance indicates that the data provide meaningful evidence against the null hypothesis, though it does not measure the practical importance of the effect.

### Example: Data from a Theoretical Distribution

Consider a scenario where the data are generated from two normal distributions. The first distribution has a mean of $\mu_1=232$ and the second has a mean of $\mu_2=219$, while both distributions share the same standard deviation, $\sigma=12$.

As a practical example, these distributions could represent the average daily ad viewing time (in minutes) for users on YouTube versus Vimeo. Suppose we have collected 90 days of viewing data for each platform. This setup allows us to compare the two distributions and explore whether the difference in means is statistically significant, providing a foundation for hypothesis testing.

````python
np.random.seed(42)

youtube = np.random.normal(232, 12, 90)
vimeo = np.random.normal(219, 12, 90)

kde_1 = sp.stats.gaussian_kde(youtube, bw_method=0.5)
kde_2 = sp.stats.gaussian_kde(vimeo, bw_method=0.5)

# Estimate kernel density for the support (x-axis)
x_values = np.arange(160, 280, 0.25)
x_densities_1 = kde_1.evaluate(x_values)
x_densities_2 = kde_2.evaluate(x_values)

plt.figure(figsize=(8, 4))
plt.plot(x_values, x_densities_1)
plt.plot(x_values, x_densities_2)
plt.show()
````

````python
plt.figure(figsize=(8, 4))
plt.hist(youtube, edgecolor='k', linewidth=1.2, bins=11, color='r', alpha=0.3, label="YouTube")
plt.hist(vimeo, edgecolor='k', linewidth=1.2, bins=11, color='b', alpha=0.3, label="Vimeo")
plt.legend()
plt.xlabel("Minutes of viewing");
````

### Assessing the Variability of Our Statistic: Different Samples

The difference between the means of two samples serves as a useful statistic for quantifying the difference between the groups. When this difference is calculated from the observed data, we refer to it as the **observed statistic**.

````python
np.mean(youtube) - np.mean(vimeo)
````

Because the data are based on random samples, collecting a new set of data will almost certainly produce a different observed statistic. This inherent variability highlights the need to account for sampling variation when interpreting results.

To properly assess whether the observed difference is meaningful, we need a distribution of the difference between the two samples. This distribution, known as the sampling distribution of the test statistic under the null hypothesis, represents the range of differences we might expect purely by chance and and provides a basis for comparing the observed statistic to what would be expected if the null hypothesis were true.

To illustrate this concept, we begin with an example where the two samples are clearly different when visualized. For this demonstration, we generate a dataset with a small standard deviation ($\sigma=2$), which makes the differences between the samples more apparent.

````python
np.random.seed(42)

youtube = np.random.normal(232, 2, 90)
vimeo = np.random.normal(219, 2, 90)

plt.figure(figsize=(8, 4))
plt.hist(youtube, edgecolor='k', linewidth=1.2, bins=11, color='r', alpha=0.5, label="YouTube")
plt.hist(vimeo, edgecolor='k', linewidth=1.2, bins=11, color='b', alpha=0.5, label="Vimeo")
plt.legend()
plt.xlabel("Minutes of viewing")
plt.show()
````

When two samples are drawn from the same distribution, the difference between their sample means tends to have an average value close to zero. This occurs because, if both samples come from the same underlying population, there is no systematic difference between them, and any observed differences arise purely from random sampling variability.

This property is particularly useful when testing the **null hypothesis**. Under the null hypothesis, we assume that any difference observed between the samples is due to chance alone rather than a true underlying effect. As a result, the distribution of differences generated from samples drawn from the same distribution provides a reference for what kinds of differences we might expect purely by random variation.

By comparing the observed difference between the sample means to this reference distribution, we can assess how unusual the observed difference is. This comparison allows us to evaluate how similar or dissimilar the samples are while properly accounting for the variability that arises from random sampling.

### Comparing Samples from Similar Distributions

An important question in hypothesis testing is what a plausible difference between sample means looks like when the underlying populations are similar. In practice, we often observe some difference between sample means simply due to random variation, even when the true populations are not very different. Understanding the range of plausible differences helps determine whether an observed result is unusual or consistent with normal sampling variability.

For example, suppose we want to investigate whether there is a difference in spending behavior between male and female customers on an e-commerce platform. To explore this scenario, we can generate data from two normal distributions representing the spending patterns of each group. The distributions are similar but not identical, with means $\mu_{male} = 466$ and $\mu_{female} = 450$. Both populations share the same standard deviation, which is relatively large ($\sigma = 89$). This relatively high variability means that individual spending amounts can vary widely, making it more challenging to distinguish differences between the groups based solely on sample data.

````python
np.random.seed(129)

male_spending = np.random.normal(466, 89, 230)
female_spending = np.random.normal(450, 89, 190)

diff_of_means = np.mean(male_spending) - np.mean(female_spending)
print(f"Difference between means is {diff_of_means}")
````

````python
all_data = np.concatenate([male_spending, female_spending])
means_difference = []

for _ in range(1000):
    np.random.shuffle(all_data)
    male_s = all_data[:230]
    female_s = all_data[230:]
    means_difference.append(np.mean(male_s) - np.mean(female_s))

plt.figure(figsize=(6, 3))
plt.hist(means_difference, edgecolor='k', linewidth=1.2, bins=12)
plt.scatter(np.mean(male_spending) - np.mean(female_spending), 0,
            color="r", s=40)
plt.ylim(-10, 200)
plt.show()
````

### How Surprising Is an Observed Statistic?

After generating 1,000 permuted statistics and visualizing their distribution, we can evaluate how unusual the observed statistic is by comparing it to this reference distribution (null distribution). Specifically, we count how many of the simulated statistics are **equal to or larger than the observed difference in means**.

This count helps us assess how surprising the observed result is under the assumption that both samples come from the same distribution. If many of the simulated statistics are as large as or larger than the observed statistic, then the observed difference is not particularly unusual, and we are less surprised by it.

Conversely, if only a small number of simulated statistics reach or exceed the observed value, the observed difference is unlikely to occur by chance alone. In that case, the evidence suggests that the spending patterns of males and females may not come from the same distribution.

````python
# One-sided 
sum(np.array(means_difference) >= diff_of_means)
````

````python
# Two-sided
sum(np.abs(np.array(means_difference)) >= abs(diff_of_means))
````

### The P-Value

The **p-value** is a convention used to quantify how far into the tails of a distribution an observed statistic must fall to be considered surprising under the null hypothesis. It is based on the **area in the tail of the sampling distribution** — either the upper tail, the lower tail, or both, depending on whether the test is one-sided or two-sided. Since we are working with an empirical distribution generated through simulation or permutation, the p-value can be computed directly by counting the proportion of simulated statistics that fall as extreme as the observed statistic.

More formally, the p-value is the probability, assuming the null hypothesis is true, of observing a test statistic that is equal to the value obtained from the data or even more extreme in the direction specified by the alternative hypothesis.

If the p-value is small, the observed statistic lies far from what the null hypothesis predicts. This suggests that the observed result would be unlikely to occur by chance alone, providing evidence in favor of the alternative hypothesis. A common convention is to consider results statistically significant when the p-value is less than 0.05, meaning that such an extreme outcome would occur less than about 1 time in 20 under the null hypothesis.

Closely related to the p-value is the **significance level**, denoted by $\alpha$. The significance level represents the probability of rejecting the null hypothesis when it is actually true, which is known as a Type I error. For example, choosing $\alpha=0.05$ means accepting a 5% risk of concluding that a difference exists when, in reality, there is no true difference.

### Hypothesis Testing Approach (General Algorithm)

The general strategy for hypothesis testing can be summarized in three main steps:

1. **Define and calculate the observed test statistic.** First, identify a metric (like the difference in means) that summarizes the effect you are looking for. Calculate this value from your actual data; this is your **observed statistic**. It represents the real-world "signal" you are testing.

2. **Construct the null distribution.** Create a reference model of what the test statistic would look like if the **null hypothesis were true**. Using methods like permutation (shuffling labels) or bootstrapping, you repeatedly recalculate the statistic on randomized data. This creates the **sampling distribution of the test statistic under the null hypothesis**, showing you what "random noise" looks like for your specific sample size and variability.

3. **Compare the observed statistic to the null distribution.** Evaluate where your observed statistic falls within that null distribution. By calculating the proportion of random trials that produced a value as extreme as — or more extreme than — your observed statistic, you determine the p-value. If the observed statistic is a rare outlier in this distribution, you have evidence to reject the null hypothesis.

### An Example

In this example, we simulate two independent samples representing Biki ride durations for two different neighborhoods, labeled `a` and `b`. Rather than using real observations, we generate synthetic data from a normal distribution with a mean $\mu=20$ minutes and a standard deviation $\sigma=4$ minutes. From this distribution, we randomly sample 15 ride durations for each of the neighborhoods `a` and `b`, giving a total of 30 simulated bike rides. Because both datasets are generated from the same underlying distribution, we would expect their statistical properties to be similar, though random sampling may still produce small differences.

First, we compute the mean ride duration for datasets `a` and `b` and compare them. Any difference between the two sample means is due to random sampling variability rather than a true difference in the populations. Based solely on the observed difference in means, it may be tempting to think one neighborhood has longer rides than the other. However, with such small samples, even noticeable differences can occur purely by chance. To further examine this, we visualize the distributions by plotting kernel density estimates (KDEs) for both datasets. The KDE plots show the smoothed distribution of ride durations for each neighborhood. If the curves largely overlap and have similar shapes and centers, this suggests that the two samples are not meaningfully different, supporting the idea that both came from the same population distribution.

````python
np.random.seed(0)
a = np.random.normal(20, 4, 15)
b = np.random.normal(20, 4, 15)
print(f"({np.mean(a):.4f}, {np.mean(b):.4f})")
print(f"Difference of means: {np.mean(a) - np.mean(b):.2f}")
````

````python
import scipy.stats as stats

x_axis = np.arange(0, 40, 0.5)
kde_1 = sp.stats.gaussian_kde(a, bw_method=1)
kde_2 = sp.stats.gaussian_kde(b, bw_method=1)
x_densities_1 = kde_1.evaluate(x_axis)
x_densities_2 = kde_2.evaluate(x_axis)
````

````python
plt.figure(figsize=(10, 4))
plt.plot(x_axis, x_densities_1, label="dataset a")
plt.plot(x_axis, x_densities_2, label="dataset b")
plt.legend();
````

Next, we construct a **null distribution** using a resampling (permutation) method. Under the null hypothesis, there is no true difference between neighborhoods `a` and `b`, meaning all 30 ride durations come from the same distribution. To simulate this, we repeatedly shuffle the combined dataset and randomly split it into two groups of 15 observations. For each resample, we compute the difference in means between the two groups and record the value. Repeating this process many times produces a distribution of differences in means that would be expected if the null hypothesis were true. We then visualize this distribution using a histogram and mark the originally observed difference in means on the plot.

````python
all_data = np.concatenate([a, b])
means_difference = []

for _ in range(1000):
    np.random.shuffle(all_data)
    a_subset = all_data[:15]
    b_subset = all_data[15:]
    means_difference.append(np.mean(a_subset) - np.mean(b_subset))

plt.figure(figsize=(8, 4))
plt.hist(means_difference, edgecolor='k', linewidth=1.2, bins=12)
plt.scatter(np.mean(a) - np.mean(b), 0, color="r", s=40)
plt.ylim(-10, 210)
plt.show()
````

Finally, we evaluate how plausible our observed statistic is under the null hypothesis. We do this by comparing the observed difference in means to the null distribution generated through resampling. The **p-value** is calculated as the proportion of resampled differences that are as extreme as or more extreme than the observed difference.

````python
obs_mean_diff = np.mean(a) - np.mean(b)
sum(means_difference >= obs_mean_diff)
````

````python
sum(means_difference >= obs_mean_diff) / len(means_difference) 
````

````python
both_sides = np.abs(np.array(means_difference)) >= np.abs(np.array(obs_mean_diff))
sum(both_sides) / len(means_difference)
````

Since the p-value of 0.118 (or 0.228 for a two-sided test) is relatively large, the observed difference between the sample means is not statistically significant. n hypothesis testing, a large p-value indicates that the observed difference is reasonably likely to occur under the null hypothesis — that is, when there is no true difference between the two groups. Because these p-values are well above a common significance threshold such as 0.05, we fail to reject the null hypothesis and conclude that there is insufficient evidence to suggest a meaningful difference in average ride durations between neighborhoods `a` and `b`.

This outcome is consistent with how the data were generated: both samples were drawn from the same normal distribution. Therefore, any difference between the two sample means is attributable to random sampling variability rather than a true underlying difference between neighborhoods. This example illustrates how hypothesis testing and resampling methods help quantify uncertainty and determine whether observed differences in data are statistically meaningful or simply the result of chance.

### Comparing a Sample to a Distribution Parameter

One common application of hypothesis testing is comparing a sample statistic to a known population parameter. For example, suppose the national average for the SAT math subscore is 525 with a standard deviation of 20. The question is: *does a sample of 100 students who followed your curriculum perform significantly differently from the national average?*

To answer this, we can generate bootstrap samples from the known national distribution and compute bootstrap statistics (replicates) to construct a 95% confidence interval. We then check whether our sample mean falls within that interval. If it does, we lack sufficient evidence to reject the null hypothesis — meaning the scores of our 100 students are statistically consistent with being drawn from a distribution with $\mu = 525$ and $\sigma = 20$. If it falls outside the interval, however, we have evidence that our students' performance significantly differs from the national average.
