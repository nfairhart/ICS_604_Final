## Hypothesis Testing

In statistics, it is important to distinguish between statistics and parameters. A **statistic** is a numerical measure calculated directly from a sample of data. It summarizes or describes a characteristic of the sample and is often used to estimate or make inferences about the corresponding population characteristic. In contrast, a **parameter** is a numerical value that describes some aspect of the entire population. It is a fixed value, often unknown, that could be determined if data from the complete population were available.

A **hypothesis** is an assumption or statement about a population parameter, such as the population mean or proportion. The goal of a **hypothesis test** is to evaluate evidence from sample data in order to decide between two competing hypotheses — typically a null hypothesis and an alternative hypothesis. By systematically analyzing the data, hypothesis testing provides a formal framework for determining whether observed differences or effects are statistically significant.

Hypothesis testing methods can be categorized as **parametric** or **non-parametric**. Parametric methods make assumptions about the underlying population parameters and distributions, such as assuming normality. In contrast, non-parametric methods do not rely on assumptions about the parameters or the shape of the underlying distributions, making them more flexible when the population structure is unknown or does not meet parametric assumptions.

### Hypothesis Testing Questions

In hypothesis testing, we are generally concerned with two main aspects. The first is the **magnitude of the difference between hypotheses**, often referred to as the **effect size**. This measures how large or meaningful the observed difference is, beyond simply determining whether it exists. For instance, economists have studied the effect of a universal basic income on happiness. While the results may be statistically significant, the effect size answers the practical question: *how much happier are people actually?* If the observed increase in happiness is only 1%, policymakers may question whether spending $3 trillion — an estimated annual cost — is justified.

The second aspect is the **probability associated with the measured difference**. This involves understanding the variability of the observed difference and assessing how likely it is that the difference could have arisen purely by chance. This is captured by the **p-value**, which represents the probability of observing a difference at least as extreme as the one measured, assuming the null hypothesis is true. The p-value provides a quantitative measure of the statistical evidence against the null hypothesis.

### Data in Hypothesis Testing

Data used in hypothesis testing can come from either one or two samples. In the case of a single sample, the goal is often to compare the sample to a population with a known parameter, such as the mean. For example, if a new math curriculum is implemented with a randomly selected group of students, their test scores can be compared to the national average. The hypothesis test then evaluates whether the curriculum leads to a significant improvement in math exam scores compared to the broader population.

Alternatively, hypothesis testing can involve two samples, where the objective is to determine whether there is a difference between them. For instance, one group of students may receive the math curriculum in the morning, while another group receives it in the afternoon. The test then examines whether the time of day affects learning outcomes, allowing us to assess whether performance differs significantly between the two groups.

### Terminology

In hypothesis testing, we typically evaluate **two competing hypotheses**. The first is the **null hypothesis** ($H_0$), which assumes that there is no change or effect in the system and that any observed differences are simply due to natural variability in the population. The second is the **alternative hypothesis** ($H_a$), which assumes that the differences we observe are unlikely to have occurred by chance alone. By "unlikely," we mean that such outcomes would be rare if the null hypothesis were true.

The alternative hypothesis can take one of two forms. A **single-tailed hypothesis** specifies that the effect or difference occurs in a particular direction. For example, a new teaching method might improve the average math score compared to the national average, expressed as $\mu_{new\_method} \gt \mu_{national\_average}$. Similarly, a new medication might reduce cholesterol levels compared to a control group, written as $\mu_{cholesterol\_treatment} \lt \mu_{cholesterol\_control}$.

A **two-tailed hypothesis**, on the other hand, considers differences in either direction. Here, the values are expected to differ from a given reference but without specifying the direction. For instance, if we want to test whether the rate of opioid prescriptions in Hawaii differs from the national average, the alternative hypothesis would allow for either $\mu_{opioids\_Hawaii} \lt \mu_{opioids\_nationally}$ or $\mu_{opioids\_Hawaii} \gt \mu_{opioids\_nationally}$.

### About the Hypothesis Tests

In hypothesis testing, the null hypothesis ($H_0$) and the alternative hypothesis ($H_a$) are **mutually exclusive**. For example, either there is a difference or there is no difference, but it would not make sense to define the hypotheses as “the difference is greater than 0” versus “the difference is greater than 10.” Each hypothesis represents a distinct scenario about the population.

It is important to understand that a null hypothesis can never be proven true. Failing to find evidence for the alternative hypothesis does not confirm the null; it only makes the null hypothesis plausible given the data. In such cases, we say that we **fail to reject the null hypothesis** or that the study did not provide sufficient evidence to reject it.

Finally, we describe results as **statistically significant** when the observed outcome is unlikely to have occurred by chance alone. Statistical significance indicates that the data provide meaningful evidence against the null hypothesis, though it does not measure the practical importance of the effect.

### Example: Data from a Theoretical Distribution

Consider a scenario where the data are generated from two normal distributions. The first distribution has a mean of $\mu_1=232$ and the second has a mean of $\mu_2=219$, while both distributions share the same standard deviation, $\sigma=12$.

As a practical example, these distributions could represent the average daily ad viewing time (in minutes) for users on YouTube versus Vimeo. Suppose we have collected 90 days of viewing data for each platform. This setup allows us to compare the two distributions and explore whether the difference in means is statistically significant, providing a foundation for hypothesis testing.

````python
np.random.seed(42)

youtube = np.random.normal(232, 12, 90)
vimeo = np.random.normal(219, 12, 90)

kde_1 = sp.stats.gaussian_kde(youtube, bw_method=0.5)
kde_2 = sp.stats.gaussian_kde(vimeo, bw_method=0.5)

# Estimate kernel density for the support (x-axis)
x_values = np.arange(160, 280, 0.25)
x_densities_1 = kde_1.evaluate(x_values)
x_densities_2 = kde_2.evaluate(x_values)

plt.figure(figsize=(8, 4))
plt.plot(x_values, x_densities_1)
plt.plot(x_values, x_densities_2)
plt.show()
````

````python
plt.figure(figsize=(8, 4))
plt.hist(youtube, edgecolor='k', linewidth=1.2, bins=11, color='r', alpha=0.3, label="YouTube")
plt.hist(vimeo, edgecolor='k', linewidth=1.2, bins=11, color='b', alpha=0.3, label="Vimeo")
plt.legend()
plt.xlabel("Minutes of viewing");
````

### Assessing the Variability of Our Statistic: Different Samples

The difference between the means of two samples serves as a useful statistic for quantifying the difference between the groups. When this difference is calculated from the observed data, we refer to it as the **observed statistic**.

````python
np.mean(youtube) - np.mean(vimeo)
````

Because the data are based on random samples, collecting a new set of data will almost certainly produce a different observed statistic. This inherent variability highlights the need to account for sampling variation when interpreting results.

To properly assess whether the observed difference is meaningful, we need a distribution of the difference between the two samples. This distribution, known as the sampling distribution of the test statistic under the null hypothesis, represents the range of differences we might expect purely by chance and and provides a basis for comparing the observed statistic to what would be expected if the null hypothesis were true.

To illustrate this concept, we begin with an example where the two samples are clearly different when visualized. For this demonstration, we generate a dataset with a small standard deviation ($\sigma=2$), which makes the differences between the samples more apparent.

<center><img src="https://www.dropbox.com/scl/fi/x2z0z433t71v7hc1av57n/two_dists.png?rlkey=4klaipmz1i7tzb0kjwbnvu2fw&st=b75d4wzo&dl=1" width="500"></center>

````python
np.random.seed(42)

youtube = np.random.normal(232, 2, 90)
vimeo = np.random.normal(219, 2, 90)

plt.figure(figsize=(8, 4))
plt.hist(youtube, edgecolor='k', linewidth=1.2, bins=11, color='r', alpha=0.5, label="YouTube")
plt.hist(vimeo, edgecolor='k', linewidth=1.2, bins=11, color='b', alpha=0.5, label="Vimeo")
plt.legend()
plt.xlabel("Minutes of viewing")
plt.show()
````

#### Question
What is the mean of the sampling distribution of the difference between the means?

### Assessing the Variability of Our Statistic: Similar Samples

Now, consider two samples drawn from the same distribution. For example, suppose we generate data for `youtube_1` and `youtube_2` using a normal distribution with the same mean and standard deviation:
 
```python
youtube_1 = np.random.normal(232, 2, 90)
youtube_2 = np.random.normal(232, 2, 90)
means_difference.append(np.mean(youtube_1) - np.mean(youtube_2))
```

Since both datasets come from the same distribution, the differences in their sample means arise purely from random sampling variation. On average, the difference between the means should be close to zero, reflecting the fact that there is no true underlying difference.

````python
means_difference = []

for _ in range(1000):
    youtube_1 = np.random.normal(232, 2, 90)
    youtube_2 = np.random.normal(232, 2, 90)
    means_difference.append(np.mean(youtube_1) - np.mean(youtube_2))

plt.figure(figsize=(8, 4))
plt.hist(means_difference, edgecolor='k', linewidth=1.2, bins=12)
plt.show()
````

````python
np.array(means_difference).mean()
````

Rather than generating entirely new datasets for each iteration, we can explore variability using an alternative approach. One option is to generate a single dataset in one call to `np.random.normal` and then split it into youtube_1 and youtube_2. We can further permute the generated data and randomly assign values to `youtube_1` and `youtube_2`. This approach allows us to simulate the sampling distribution of the difference in means for two samples drawn from the same distribution.

````python
# Example

a = np.array([1, 2, 3, 4, 5])

# shuffles the data in place
np.random.shuffle(a)
print(f"Shuffled data is {a}")

print(f"Chunk 1 is {a[:2]}")
print(f"Chunk 2 is {a[2:]}")
````

````python
# Generating the data in the same call, 
# shuffling and splitting the data does not change the results

means_difference = []

for _ in range(1000):
    # Generate 180 days worth of data in each iteration
    all_data = np.random.normal(232, 2, 180)
    
    # shuffle the data and then split it over youtube_1 and youtube_2
    np.random.shuffle(all_data)
    youtube_1 = all_data[0:90]
    youtube_2 = all_data[90:]
    means_difference.append(np.mean(youtube_1) - np.mean(youtube_2))

plt.figure(figsize=(8, 4))
plt.hist(means_difference, edgecolor='k', linewidth=1.2, bins=12)
plt.show()
````

````python
np.array(means_difference).mean()
````

### Original Observation Vs. Distribution Under the Null Hypothesis

````python
means_difference = []

youtube = np.random.normal(232, 2, 90)
vimeo = np.random.normal(219, 2, 360)
all_data = np.concatenate([youtube, vimeo])

for _ in range(1000):
    np.random.shuffle(all_data)
    youtube_s = all_data[:90]
    vimeo_s = all_data[90:]
    means_difference.append(np.mean(youtube_s) - np.mean(vimeo_s))

plt.figure(figsize=(8, 4))
plt.hist(means_difference, edgecolor='k', linewidth=1.2, bins=12)
plt.scatter(np.mean(youtube) - np.mean(vimeo), 2, color="r", s=40)
plt.ylim(-10, 230)
plt.show()
````

When two samples are drawn from the same distribution, the difference between their sample means tends to have an average value close to zero. This occurs because, if both samples come from the same underlying population, there is no systematic difference between them, and any observed differences arise purely from random sampling variability.

This property is particularly useful when testing the **null hypothesis**. Under the null hypothesis, we assume that any difference observed between the samples is due to chance alone rather than a true underlying effect. As a result, the distribution of differences generated from samples drawn from the same distribution provides a reference for what kinds of differences we might expect purely by random variation.

By comparing the observed difference between the sample means to this reference distribution, we can assess how unusual the observed difference is. This comparison allows us to evaluate how similar or dissimilar the samples are while properly accounting for the variability that arises from random sampling.
