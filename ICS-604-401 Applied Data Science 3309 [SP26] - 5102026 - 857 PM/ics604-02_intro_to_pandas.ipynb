## Python Modules and Packages

In Python, the concepts of modules and packages exemplify the DRY (Don’t Repeat Yourself) principle in software engineering by promoting code modularity, reuse, and maintainability. A **module** refers to an individual Python file containing a collection of related definitions and implementations — such as functions, classes, and variables — that serve a specific purpose within a program. This approach enables the logical organization of code and facilitates its reuse across multiple programs or projects. For example, a file named `math_operations.py` containing a suite of mathematical function definitions would constitute a typical module.

In practice, it is often impractical to combine all the desired functionality into a single module (i.e., a single Python file). For example, if you work in analytics for Hawaiian Airlines, you might separate customer analytics and freight analytics into two different modules. For organizational convenience, both of these modules can be placed within the same folder.

A **package** in Python is essentially a directory that contains a collection of related modules. A package directory typically includes an `__init__.py` file, which designates the directory as a Python package and enables its constituent modules to be properly imported. In addition to modules, a package can also contain other files such as documentation or configuration files. Packages can be further organized into subpackages to handle more specific divisions of functionality. For instance, the freight analytics package could be split into separate modules for commercial freight and personal freight.

A package establishes a shared top-level namespace for the modules it contains, while each module retains its own nested namespace. This hierarchical arrangement helps prevent naming conflicts and supports the systematic management of large codebases, facilitating the development of complex software systems. The top-level package in such a hierarchy is often referred to as a library, providing a structured collection of modules and subpackages that together implement a broad set of functionality.

Modules are imported using the `import <module>` statement, whereas the functionalities within packages are accessed through their respective module names by means of **dot notation** (e.g., package_name.module_name).

### Importing Python Modules

To use functionality defined in a Python module, the module must first be imported into your program. Importing makes the module’s definitions—such as variables, functions, and classes—available for use within your file.

One common way to import a module is to import the entire module while keeping its functionality bound to the module name. This is done using the `import <module>` statement. For example,

```python
>>> import math
```

imports the built-in `math` module and makes all of its definitions accessible through the module name. After this import, you access items in the module using dot notation. For instance, to access the mathematical constant *$\pi$* defined in the math module, you would use:

```python
>>> math.pi
3.141592653589793
```

Similarly, to compute the factorial of 4 using the `math` module’s `factorial` function, you would type:

```python
>>> math.factorial(4)
24
```

This style of importing helps keep code organized and avoids naming conflicts by clearly indicating the source of each function or value.


Another way to import functionality from a Python module is to import only specific names rather than the entire module. This approach allows you to access selected definitions directly, without prefixing them with the module name.

For example, if you only want to import the mathematical constant *$\pi$* from the `math` module, you can use the statement:

```python
>>> from math import pi
```

After doing so, the name `pi` is available directly in your namespace, and you can reference it simply by typing `pi`:

```python
>>> pi
3.141592653589793
```

This style of importing can make code more concise, but it should be used carefully to avoid name conflicts, since the imported names are no longer explicitly associated with the module they came from.

````python
import math

math.pi
````

````python
# Computes n!
math.factorial(4)
````

#### Importing Python Modules — What *Not* to Do

When importing Python modules, it is generally best to **avoid using the `*` wildcard** to import all functionality from a module. For example, writing `from math import *` brings every public name defined in the `math` module into your current workspace (namespace). As a result, functions and constants such as `pi`, `sqrt`, and `sin` can be accessed directly without the `math.` prefix, as in `c = 2 * pi * r`.

Although this may seem convenient, wildcard imports have several drawbacks. First, they **pollute the namespace** by introducing many names whose origins are unclear, making it difficult to know where a particular function or variable comes from. Second, they **create the potential for name clashes**, where imported names may unintentionally overwrite existing variables or functions, leading to subtle and hard-to-detect bugs. Finally, code that relies on wildcard imports is **harder to read and maintain**, since it is not immediately obvious which module provides a given piece of functionality.

For these reasons, it is considered best practice to import modules explicitly (for example, `import math` or `from math import pi`) to keep code clear, predictable, and easier to understand.

### Providing a Package Alias

When importing Python modules or packages, it is common practice to assign an alias to the imported name. Aliasing allows you to use a shorter, more convenient name in your code, saving keystrokes and improving readability—especially for modules with long names.

For example, the `pyplot` submodule from `matplotlib` is frequently imported using an alias:

```python
from matplotlib import pyplot as plt
```

With this alias, all functionality provided by `pyplot` can be accessed using `plt` instead of the longer `pyplot` name:

```python
plt.scatter(x, y)
```

Using aliases does not change the functionality of the module; it simply provides an alternative name within your namespace. In practice, the expression `from X import Y as Z` is often replaced by the clearer and more common form `import X.Y as Z`, as shown in the example below:

```python
import matplotlib.pyplot as plt
```

Aliases are typically standardized by convention and, in many cases, recommended by the package developers themselves. Common examples include:

- `pd` for `Pandas`
- `np` for `NumPy`
- `sns` for `Seaborn`
- `plt` for `pyplot`

Following these conventions makes code easier to read and understand, especially when collaborating with others or reading existing codebases.

### Types of Python Modules and Packages

Python modules and packages can be broadly grouped based on how they are distributed and installed. The first category is the Python standard library, which is distributed with Python itself. These modules are available immediately after installing Python and do not require any additional installation, although they must still be explicitly imported before use. Examples include `datetime`, `random`, and `collections`. The standard library provides core functionality that supports a wide range of common programming tasks.

The second category consists of third-party packages commonly included in popular Python distributions, such as Anaconda or Conda-based environments. These packages are not part of the standard library, but they are often pre-installed because of their widespread use, especially in data science and scientific computing. Examples include `numpy`, `pandas`, and `matplotlib`. While these packages may appear to be available "by default" in such environments, they are still third-party modules and must be imported explicitly. In minimal Python or minimal Conda installations, they may need to be installed manually.

The third category includes specialized third-party packages that are not included by default in most Python distributions and must be installed explicitly by the user. These packages are typically domain-specific and are used for specialized tasks, such as bioinformatics, astronomy, or machine learning. Examples include `biopython`, `astropy`, and `tensorflow`. Technically, these packages are no different from the commonly bundled third-party packages; the distinction lies primarily in their scope, audience, and whether they are pre-installed for convenience.

### `Series`

A **Pandas Series** is conceptually similar to a Python list, but with important enhancements that make it more suitable for data analysis. Like a list, a Series represents an **ordered collection of items**, meaning that the order of the data is preserved. Unlike a plain list, however, a Series is designed to work efficiently with numerical and structured data and supports many vectorized operations.

All values in a Series share a **single underlying data type**. If values of different types are provided, Pandas will automatically convert them to a common type —- often the `object` data type —-to maintain this consistency. While convenient, this automatic type conversion can be a common source of bugs or unexpected behavior, especially when numerical operations are intended but the data has been silently converted to objects.

Each element in a Series is associated with a **label**, known as the **index**. These labels provide meaningful identifiers for the data and allow values to be accessed by label in addition to their positional order. The presence of an index distinguishes a Series from a simple list and enables powerful alignment and selection features in Pandas.

#### Example of Series as rows and columns

It is often helpful to think of a Pandas Series as representing a single row or a single column in an Excel worksheet, where the index corresponds to the row or column labels. This analogy makes it easier to understand how Series fit into the broader structure of Pandas DataFrames, which are built from multiple Series aligned by their indices.

### Indexing Series

A Pandas Series supports flexible and powerful indexing mechanisms that combine features of both Python lists and dictionaries. You can access elements **by position** using the `.iloc` (index location) operator, which works with integer positions. For example, `s.iloc[1]` retrieves the second element in the Series `s`, while `s.iloc[-1]` accesses the last element.

In addition to positional indexing, you can access values in a Series **by label** using its index. For example, if a Series `s` has labels such as `"medication"` or `"spending"`, you can retrieve the corresponding values using the indexing operator with square brackets, e.g., `s["medication"]` or `s["spending"]`, or by using the `.loc` operator, e.g., `s.loc["medication"]` or `s.loc["spending"]`. Using label-based access makes your code more readable and allows you to work with data using meaningful identifiers rather than relying solely on positions.

Because a Series allows access both by integer position (like a list) and by label (like a dictionary), it is fair to think of a Pandas Series as a hybrid between a list and a dictionary. This dual nature is one of the key features that makes Series a powerful and flexible data structure for data analysis.

````python
import pandas as pd

s = pd.Series(data = [1234, 'DIAZEPAM', 3, '$32'], 
              index = ['doctor_id', 'medication', 'nb_beneficiaries', 'spending'])
s
````

````python
print(s.iloc[1], s.iloc[-1])
````

````python
print(s["medication"], s["spending"])
````

````python
print(s.loc["medication"], s.loc["spending"])
````

````python
# This code raises a deprecation warning 

print(s[1])     # single number is interpreted as a label by default
print(s["medication"])
````

### Subsetting Series 

A Pandas Series allows you to extract subsets of elements in several flexible ways. One common approach is to use the `.iloc` operator with a list of integer positions. For example, `s.iloc[[0, 2, 1]]` returns multiple elements from the Series in the specified order. In this expression, the outer square brackets apply the indexing operation, while the inner square brackets define the Python list of positions to select.

Subsetting can also be performed using slice notation with the range operator (`:`). For example, `s[0:3]` or `s.iloc[0:3]` returns the first three elements of the Series. As with Python lists, the starting index is inclusive and the ending index is exclusive. This method is convenient when you want to select a contiguous block of elements based on their positions.

````python
s.iloc[[0, 2, 1]]
````

````python
s[0:3]     # numbers are treated as positions in slicing without .loc
````

````python
s.iloc[0:3]
````

In addition to positional indexing, you can subset a Series using a list of index labels. For example, `s[["doctor_id", "nb_beneficiaries"]]` or `s.loc[["doctor_id", "nb_beneficiaries"]]` selects only the entries with those labels. Here again, the outer brackets perform the indexing, and the inner brackets define the list of labels to select. You can also use slice notation with labels, making it easy to select a continuous range of elements by their index names.

````python
s[["doctor_id", "nb_beneficiaries"]]
````

````python
s["doctor_id": "nb_beneficiaries"]
````

````python
s.loc[["doctor_id", "nb_beneficiaries"]]
````

````python
s.loc["doctor_id": "nb_beneficiaries"]
````

### Index Location

In a DataFrame, rows can be indexed by position using the `.iloc` (index location) operator. You can pass `.iloc` either a single integer index or a list of integer indexes. Unlike methods or functions that use parentheses `()`, `.iloc` uses square brackets `[]`.

For example:

```python    
>>> spending_df.iloc[3]        # single row 
>>> spending_df.iloc[[1, 5]]   # multiple rows
```

When a single index is provided, Pandas returns a Series representing that row, since each row in a DataFrame is essentially a Series. When a list of indexes is provided, Pandas returns a subset DataFrame containing only the specified rows.

````python
# Returns row, or Series

spending_df.iloc[3]
````

````python
# Returns two rows as a DataFrame

spending_df.iloc[[1, 5]]
````

````python
spending_df.iloc[1:4]
````

### Indexing Rows and Columns with `.iloc` 

The .iloc operator can also be used to select a combination of rows and columns from a DataFrame. The general syntax is:

```python
 data_frame.iloc[row_index_info, column_index_info]
```

Here, `row_index_info` and `column_index_info` can each be either a single integer or a list of integers. This allows you to extract a single value, a row, a column, or a subset of the DataFrame by specifying the positions of the rows and columns you want. For example, `spending_df.iloc[3, 3]` returns the value at the fourth row and fourth column, while `spending_df.iloc[[2, 4], [1, 3]]` returns a smaller DataFrame containing the third and fifth rows and the second and fourth columns.


````python
spending_df.iloc[3, 3]
````

````python
spending_df.iloc[2, [1, 3]]
````

````python
spending_df.iloc[[2, 4], [1, 3]]
````

````python
spending_df.iloc[:, [1, 3]]
````

### Subsetting by Range on Labels

In addition to selecting rows and columns by individual labels, Pandas also allows you to use range (slice) notation with labels when indexing a DataFrame using `.loc`. Although this may not always be intuitive or commonly used, it is fully supported for both row indexes and column labels.

For example, the following expressions are all valid:

```python    
>>> spending_df.loc["AB789982", "specialty":"spending"]    
>>> spending_df.loc["AB789982":"GZ129032", "specialty"]    
>>> spending_df.loc["AB789982":"GZ129032", "specialty":"nb_beneficiaries"]    
```

When slicing with labels using `.loc`, **both the start and end labels are included** in the result. This behavior differs from Python list slicing, where the upper bound is excluded. Label-based slicing can be useful when working with ordered or meaningful labels, but should be used carefully to avoid confusion.


````python
spending_df.loc["AB789982":"GZ129032", "specialty":"nb_beneficiaries"]
````

| Syntax                |   Meaning    |
|:-------------------------------------|:-------------------|

| DataFrame.iloc[ [0, 2] ] | returns entries of indices 0 and 2 |
| DataFrame.iloc[23, [0, 1] ] | returns entry of index 23, and only values of columns 0 and 1 |

| DataFrame[ ["col_name_1", "col_name_2"] ]    |  Returns `DataFrame` with  "col_name_1" and "col_name_2" |
| DataFrame.loc[:, ["col_name_1", "col_name_2"] ]    |  Returns `DataFrame` with  "col_name_1" and "col_name_2" |
| DataFrame.loc["label"] | returns entry indexed by "label" |
| DataFrame.loc["label", ["col_3", 'col_5'] ] | returns entry indexed by "label", subsets columns to only "col_3" and "col_5" |
| DataFrame.loc[ ["label_1", "label_2"], ['col_3', 'col_5'] ] | returns entries with indices "label_1" and "label_2", subsets columns to only "col_3" and "col_5" |
