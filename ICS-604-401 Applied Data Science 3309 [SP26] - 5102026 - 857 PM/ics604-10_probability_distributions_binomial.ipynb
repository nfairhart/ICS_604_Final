## Probability Models

Probability models provide a structured way to reason about systems that involve uncertainty. They allow us to analyze a probabilistic system from multiple perspectives—what we expect to happen on average, how much outcomes might fluctuate, and how likely specific events are to occur.

Consider running an advertising campaign through an agency. You might first ask: *What is the average number of new clients we should expect after the campaign?* This reflects the expected value of the outcome. However, the average alone is not enough. You would also want to determine how much variability exists in the system. For example, expecting between 300 and 400 new clients provides much clearer guidance than a wide range such as 100 to 1,000. Understanding variability helps quantify risk and uncertainty. Finally, you may want to compute the probability of specific events — such as the probability of acquiring more than X new clients after the campaign. This supports decision-making under uncertainty.

Probabilistic models are built around three core components. First, **random variables** represent what is uncertain in the system — in this case, the number of new clients generated by the campaign. Second, **probability distributions** describe how that uncertainty behaves by assigning likelihoods to different possible outcomes. Finally, **parameters** are the numerical values that specify the exact form of the distribution, determining its center, spread, and overall shape. Together, these components allow us to formally model, analyze, and reason about uncertain outcomes.

### Probability Models: An Example

Consider an advertising campaign run through an agency, where you are interested in three related questions: the average number of new clients generated by the campaign, the variability around that average, and the probability of obtaining at least $X$ new participants. These questions reflect the core components of a probability model: summary behavior, variability, and event probabilities.

One way to address these questions is to repeat the ad campaign a large number of times, either in reality or through simulation, and record the number of new participants observed in each run. This produces a collection of outcomes, such as:

```python
    nb_new_participants = [500, 800, 860, 820, 840, 760, 430,...]
```

From this data, we can compute the quantities of interest directly. The average number of new clients is estimated using the sample mean:

```python
    np.mean(nb_new_participants)
```

The variability in the number of new clients across campaigns is captured by the variance:

```python
    np.var(nb_new_participants)
```

Finally, the probability of observing at least $X$ new participants can be estimated by computing the proportion of campaign runs in which the number of participants exceeds $X$:

```python
    sum(array(nb_new_participants) > X) / len(nb_new_participants)
```

The same information is also conveyed through a visualization of the data, such as a histogram. Each bar represents how frequently a particular range of outcomes occurs across repeated campaigns. From the shape and spread of the histogram, we can visually infer the mean, assess variability, and estimate the probability that the number of new participants exceeds a chosen threshold $X$. Visualization thus provides an intuitive complement to numerical summaries when working with probability models.

### Probability Distributions

At the core of probability models are **random variables** and **probability distributions**. A random variable represents the outcome of a random experiment in numerical form, while a probability distribution describes how likely each possible value of that random variable is.

Consider a loaded die, where the outcomes 1, 2, 3, 4, 5, and 6 do not occur with equal probability. Suppose the probabilities of each face are given by:

```
    p(1) = 0.3/6
    p(2) = 0.7/6
    p(3) =   2/6
    p(4) = 0.5/6
    p(5) = 0.2/6
    p(6) = 2.3/6
```

We can view these probabilities as the values returned by a function that takes an outcome from the sample space `(1, 2, 3, 4, 5, 6)` as input and returns the corresponding probability from the list `[0.3/6, 0.7/6, 2/6, 0.5/6, 0.2/6, 2.3/6]`. Such a function is called a probability distribution.

In essence, a probability distribution is a numerical function that assigns probabilities — values between 0 and 1 on the y-axis — to outcomes in the sample space on the x-axis. The sample space forms the domain of the function and may be numerical (as in the case of a die) or non-numerical (such as categories or labels). Probability distributions provide the fundamental link between random processes and quantitative reasoning about uncertainty.

````python
def loaded_die_distribution(face):
    try:
        dist = {1: 0.3/6, 2: 0.7/6, 3: 2/6, 4: 0.5/6, 5: 0.2/6, 6: 2.3/6}
        return dist[face]
    except:
        print(f"{face} is not a valid die face")
        raise
````

````python
sample_space = list(range(1, 7))
p_x = [loaded_die_distribution(f) for f in sample_space]

plt.figure(figsize=(8, 4))
plt.bar(sample_space, p_x)

plt.xlabel('Die face obtained')
plt.ylabel('probability');
````

### Sampling from the Biased Die

Given the probability distribution for the loaded die, some outcomes are more likely than others. For example, rolling a 3 is four times more likely than rolling a 4, reflecting the unequal probabilities assigned to each face. To simulate rolls of this biased die, we must encode the probability distribution so that sampling reflects these differences.

This can be done directly using NumPy’s `np.random.choice()` function, which allows us to specify both the set of possible outcomes and their associated probabilities. By providing the probability distribution through the `p` parameter, each draw respects the bias built into the die:

```python
np.random.choice([1, 2, 3, 4, 5, 6], p=[0.3/6, 0.7/6, 2/6, 0.5/6, 0.2/6, 2.3/6])
```

````python
for _ in range(20):
    print(np.random.choice([1, 2, 3, 4, 5, 6], 
                           p=[0.3/6, 0.7/6, 2/6, 0.5/6, 0.2/6, 2.3/6]
                          ), 
          end=' ')
````

### Random Variables

From a computer science perspective, a variable is simply a container whose value may change over time. A random variable extends this idea by allowing the value assigned to the variable to be determined by a random process. In probabilistic thinking, a random variable can take any value from the sample space of an experiment, with the specific value governed by an underlying probability distribution.

For example, when we simulate a roll of a biased die, the value assigned to a variable depends on a random draw:
  
  ```python
      # randomly select a value according to a probability distribution
      dieRollVal = np.random.choice([1, 2, 3, 4, 5, 6], 
                                    p=[0.3/6, 0.7/6, 2/6, 0.5/6, 0.2/6, 2.3/6]);
  ```
  
Formally, a random variable is defined as a **function from the sample space of an experiment to the set of real numbers**. That is, it assigns a numerical value to each possible outcome of the random experiment. The randomness lies not in the function itself, but in which outcome of the experiment is realized.

In a non-probabilistic program, the value stored in a variable can be traced deterministically from the code. In a probabilistic program, this is no longer possible: the value of a random variable is not known before the code executes. What we can reason about, however, is the distribution of possible values the variable may take and which values are more or less likely.

Consider the variable `roll_random_var` in the following loop:

````python
for i in range(3):
    roll_random_var = np.random.choice([1, 2, 3, 4, 5, 6], 
                                       p=[0.3/6, 0.7/6, 2/6, 0.5/6, 0.4/6, 2.1/6])
    print(f"roll_random_var contains the value {roll_random_var}")
````

Here, `roll_random_var` is a random variable because the value assigned to it at each iteration is determined probabilistically using the probability distribution of the biased die. Each execution may produce a different value, but the long-run behavior of the variable is fully described by its probability distribution.

````python
from collections import Counter

roll_random_var_list = np.random.choice([1, 2, 3, 4, 5, 6], size=10000,
                                        p=[0.3/6, 0.7/6, 2/6, 0.5/6, 0.4/6, 2.1/6])

counts = Counter(roll_random_var_list)
sorted_counts = {int(k): v for k, v in sorted(counts.items())}
print(sorted_counts)

plt.figure(figsize=(12, 4))
plt.bar(sorted_counts.keys(), sorted_counts.values())
plt.show()
````

### Two Types of Random Variables

Random variables are commonly classified into two broad categories: **discrete** and **continuous**, based on the nature of the values they can take.

A discrete random variable takes values from a finite or countably infinite set, typically integers, and assigns a probability to each possible outcome. Examples include the number of accidents on the H1 on a given day, the number of customers at a campus Starbucks during lunchtime, or the number of siblings a student at UH has. In each case, the possible outcomes can be listed or counted, even if the list is infinite.

In contrast, a continuous random variable takes values from an uncountably infinite set, usually represented by real numbers. Instead of assigning probabilities to individual values, continuous random variables are described using a probability density function, which indicates how dense outcomes are around a given value. Examples include height, weight, and time, where the sample space is infinite and individual exact values have probability zero.

Both discrete and continuous random variables are associated with commonly occurring probability distributions. These distributions are simply mathematical functions — like the ones used earlier — that describe how probability or probability density is distributed across the sample space. Among the most widely used distributions are the **binomial distribution** for discrete random variables and the **normal (or Gaussian) distribution** for continuous random variables, which appear frequently in data science and real-world modeling.

## The Binomial Distribution (The `"Hello World"` of Probability Distribution)

The binomial distribution models the probability of observing a certain number of *successes* in a fixed number of repeated trials. Specifically, it describes the probability of obtaining exactly $m$ successes in an experiment consisting of $n$ trials, where each trial is conducted under identical conditions and is independent of all others.

Each trial has only two possible outcomes — commonly labeled *success* and *failure*. The probability of success is denoted by $p$, while the probability of failure is $1−p$. Classic examples include flipping a coin (heads vs. tails), recording whether a customer makes a purchase (yes vs. no), or checking whether an email recipient volunteers (1 vs. 0). Because of its simplicity and wide applicability, the binomial distribution is often the first probability distribution encountered in probability and statistics and serves as a foundational building block for more advanced models.

###  Examples of the Binomial Distribution

Each of the following scenarios can be modeled using a binomial distribution because they all involve a fixed number of independent trials, each with two possible outcomes and a constant probability of success.

- A fair coin is flipped 100 times, and we are interested in the probability of observing exactly 27 heads. Here, each coin flip is a trial with two outcomes (heads or tails). The probability of success (heads) is $p=0.5$, the number of trials is $n=100$, and the number of successes of interest is $m=27$.

- An online advertisement has a clickthrough probability of $p=0.1$, and the ad is shown 1,000 times. Each ad impression represents an independent trial with two outcomes: click or no click. The probability of observing exactly 100 clicks corresponds to $n=1,000$ trials, $p=0.1$, and $m=100$ successes.

- A disease has a prevalence probability of $p=0.05$ in a population. If 5,000 individuals are randomly selected and tested, each test constitutes a trial with two outcomes: disease present or not. Here, $n=5,000$, $p=0.05$, and the event of interest is observing at least 100 cases, meaning $m \geq 100$. Unlike the previous examples, this question asks for the probability of a range of outcomes rather than a single exact count.

Across all these cases, $n$ represents the total number of trials, $p$ the probability of success on any single trial, and $m$ the number of successes whose probability we want to compute. Identifying these three components is the key step in recognizing when a problem can be modeled using the binomial distribution.

### Simple Probability Distribution

Let $X$ be the random variable representing the *number of heads* obtained when flipping a fair coin five times. Each coin flip is an independent trial with two possible outcomes: heads (H) or tails (T). The experiment therefore consists of five independent binary outcomes.

The *sample space* of this experiment is the set of all possible sequences of heads and tails that can occur in five coin flips. Formally, this sample space is the Cartesian product:

$$
[H,T]^5 = [H,T] \times [H,T] \times [H,T] \times [H,T] \times [H,T]
$$

which contains $2^5=32$ distinct outcomes. Each outcome corresponds to a specific ordered sequence, such as $HTHTH$ or $TTTTT$.

The random variable $X$ does not distinguish between different sequences with the same number of heads. Instead, it maps each outcome in the sample space to an integer value representing how many heads appear in that sequence. As a result, while the sample space consists of 32 possible outcomes, the random variable $X$ can take only the values {0, 1, 2, 3, 4, 5}. This mapping from outcomes to numerical values is what allows us to define a probability distribution for $X$.

````python
for p in itertools.product(['H', 'T'], repeat=2):
    print(p)
````

### Manually Deriving the Probability Distribution

We are interested in deriving the probability distribution of the random variable $X$, where $X$ denotes the number of heads obtained when flipping a fair coin five times. Because the experiment consists of a fixed number of independent trials with two possible outcomes and a constant probability of success, $X$ follows a binomial distribution with parameters $n=5$ and $p=0.5$.

The random variable $X$ can take the values {0, 1, 2, 3, 4, 5}, corresponding to the possible numbers of heads observed. Although there are only six possible values of $X$, the underlying sample space contains 32 equally likely outcomes, each corresponding to a distinct sequence of heads and tails.

We can compute the probability of each value of $X$ by counting how many sequences lead to that outcome:

* There is exactly one sequence with zero heads: $TTTTT$ <br/>
     $p(X=0) = 1/32$
     
* There are five sequences with exactly one head: $HTTTT, THTTT, TTHTT, TTTHT, TTTTH$.  <br/>
     $p(X=1) = 5/32$
     
* There are ten sequences with exactly two heads, such as: $HHTTT, HTHTT, HTTHT, HTTTH, …, TTTHH$. <br/>
     $p(X=2) = 10/32$
     
The same counting logic applies for three, four, and five heads.

In general, the number of distinct sequences that produce exactly $m$ heads in $n$ coin flips is given by the **binomial coefficient**: ${n}\choose{m}$ (read as $n$ choose $m$). This quantity counts the number of ways to choose which $m$ of the $n$ flips result in heads and is defined as
$$
{{n}\choose{m}} = \frac{n!}{m!(n-m)!}
$$

For example, when $n=5$ and $m=2$,
\begin{equation} 
\begin{split}
{{5}\choose{2}} & = \frac{5!}{2!(5-2)!} \\
                & = \frac{5\times4\times3\times2 \times 1}{2 \times 1 \times 3 \times 2 \times 1} \\
                & = \frac{5 \times 4}{2 \times 1} \\
                & = 10 
\end{split}
\end{equation}

Because all 32 outcomes are equally likely, the probability of observing exactly $m$ heads is simply the number of favorable outcomes divided by 32. This reasoning leads directly to the general binomial probability formula.

````python
X = [0, 1, 2, 3, 4, 5]
p_x = [1/32, 5/32, 10/32, 10/32, 5/32, 1/32]

plt.figure(figsize=(10,4))
plt.bar(X, p_x, color="#3498db")
plt.xlabel('Number of heads in five coin flips')
plt.ylabel('Probability')
plt.show()
````

### Binomial Distribution with Biased Coin

So far, we have assumed a fair coin. However, the binomial framework also applies when the coin is *biased*, as long as each flip is independent and the probability of heads remains constant. Suppose the probability of heads is $p(H) = 0.7$ and $p(T) = 0.3$.

We are interested in computing the probability of obtaining *exactly one head* in five coin flips.

#### Bernoulli Distribution (Single Trial)

Before addressing multiple flips, recall from elementary statistics and discrete math the *Bernoulli distribution*, which models the outcome of a single trial with two possible outcomes. Its probability mass function is

$$
 \Pr(k;p)=p^{k}(1-p)^{1-k}\quad {\text{for }}k\in \{0,1\}
$$

where
* $k=1$ represents success (head),
* $k=0$ represents failure (tail),
* $p$ is the probability of success.

From this formula:

\begin{align}
\Pr(1; p) &= p^1 (1-p)^0 = p \\
\Pr(0; p) &= p^0 (1-p)^1 = 1-p
\end{align}

This matches our intuition: a biased coin lands heads with probability $p$ and tails with probability $1−p$.

#### From Bernoulli to Binomial

Flipping a coin five times corresponds to performing five independent Bernoulli trials. The number of heads observed, denoted by the random variable $X$, therefore follows a binomial distribution with parameters $n=5$ and $p=0.7$.

To obtain exactly one head, we must:
1. Choose which one of the five flips results in a head, and
2. Compute the probability of that specific sequence.<br>

There are ${{5}\choose{1}}=5$ distinct sequences with exactly one head, such as: $HTTTT, THTTT, TTHTT, TTTHT, TTTTH$. Each of these sequences has probability $(0.7)^1(0.3)^4$. Therefore, the total probability of observing exactly one head is

$$
p(X=1) = {{5}\choose{1}}p(H)^1p(T)^4 = 5 \times 0.7 \times 0.3^4 = 0.02835.
$$

This example illustrates how the binomial distribution naturally arises from repeated Bernoulli trials and how it accommodates both fair and biased coins.

### The Binomial Distribution (continued)

The binomial probability formula derived above applies to any number of trials $n$ and any probability of success $p$. Writing the probability of heads as $p$ and the probability of tails as $1−p$, the probability that the random variable $X$ takes the value $m$ is given by
$$p(X=m) = {{n}\choose{m}} p^m (1-p)^{n-m}$$

This expression is the **probability mass function** of a binomially distributed random variable $X$. It provides a compact and systematic way to assign probabilities to every possible outcome in the sample space $\{ 0, 1, …, n \}$.

Having access to this distribution allows us to reason analytically about real-world problems that naturally fit the binomial framework. For example, suppose the probability that a user subscribes to a monthly plan in your app is $p=0.05$, and on average 10,000 users download the app each month. Over one year, this results in a fixed number of independent subscription opportunities, each with the same probability of success.

Using the binomial distribution, we can compute:
- the *expected number* of subscribed users after one year,
- the *variability* around that expectation,
- and the *probability* of observing at least a given number of subscribers (e.g., 3,000).

These quantities provide actionable insights for decision-making, such as estimating total revenue, planning server capacity, and determining staffing needs for customer support. The binomial distribution thus serves as a simple yet powerful probability model for reasoning about repeated, independent events with binary outcomes.

````python
from scipy.special import comb

def binomial_probability(n, m, p):
    """
    Calculate the binomial probability.

    :param n: Number of trials
    :param m: Number of successful outcomes
    :param p: Probability of success in each trial
    :return: Binomial probability
    """
    probability = comb(n, m) * (p ** m) * ((1 - p) ** (n - m))
    return probability

n = 10  # Total number of trials
m = 3   # Number of successes
p = 0.5 # Probability of success in each trial

print(binomial_probability(n, m, p))
````

Once the parameters of a probability distribution are known — such as the subscription probability $p=0.05$ in the example above — it becomes straightforward to answer many practical questions, such as those listed earlier, using the model. In some cases, these computations are trivial; in others, they may require more careful mathematical or computational treatment.

However, this raises a deeper and more important question: *where do these parameters come from?* In practice, values like $p=0.05$ are rarely known with certainty. They are typically estimated from data, prior experiments, historical observations, or assumptions about user behavior. As a result, there is inherent uncertainty in the parameter itself. How confident are we that $p=0.05$ is accurate? How plausible is it that the true value is $p=0.06$ or $p=0.04$ instead?

These questions highlight the distinction between specifying a probability model and **inferring its parameters** from data — a transition that naturally leads from probability theory to statistics and statistical inference.

## Probability Distributions for Discrete Random Variables

A probability distribution associated with a **discrete random variable** is called a **probability mass function (pmf)**. A pmf maps each possible value of a discrete random variable $X$ to a probability, assigning a likelihood to every element of the variable’s sample space.

For many canonical discrete distributions, such as the binomial distribution, the pmf can be derived directly from first principles — that is, from the basic definitions of probability. This involves explicitly defining the underlying experiment, specifying assumptions such as independence and fixed success probabilities, enumerating the possible outcomes, and computing probabilities by counting how many outcomes correspond to each value of $X$. The binomial pmf, for example, is obtained by counting the number of ways a fixed number of successes can occur across repeated independent trials and multiplying by the probability of each such outcome.

Once these probability mass functions are known, they provide a powerful and convenient way to compute probabilities of events. Beyond evaluating individual probabilities, pmfs allow us to compute important summary quantities such as expected values and variances, evaluate probabilities over subsets of the sample space, and generate random samples from the distribution. For these reasons, pmfs are central tools in both theoretical probability and practical data analysis.

### Probability Mass Functions (`pmf`) in Python

Modern programming languages and data analysis environments provide built-in support for commonly used probability mass functions, making it straightforward to work with discrete random variables in practice. In Python, this functionality is primarily available through scientific computing libraries such as NumPy and SciPy, which together form the foundation of the scientific Python ecosystem.

NumPy provides the core infrastructure, including high-performance multidimensional arrays and efficient tools for numerical computation and random number generation. These capabilities make NumPy well suited for simulating random experiments, storing large samples, and performing basic probability calculations.

SciPy builds on NumPy by offering a rich collection of higher-level scientific functions, including implementations of many standard probability distributions, statistical analysis tools, optimization routines, and numerical integration methods. SciPy is therefore especially useful when working directly with probability distributions, computing pmfs analytically, or performing more advanced statistical inference.

Together, NumPy and SciPy provide a powerful and flexible framework for computing, visualizing, and reasoning about probability mass functions in practical data analysis workflows. A common way to visualize a pmf is with a histogram-style plot, where the x-axis shows the possible observed values and the y-axis represents the corresponding probabilities. Each bar indicates the probability mass at a particular outcome, and the total probability across all values sums to one.

````python
from scipy.stats import binom

n = 5 
p = 0.7
print(binom.pmf(1, n, p))
````

### Binomial Distribution Summary: `X`, `n`, `p`, and the `pmf` 

In a binomial experiment, the number of observed successes is treated as a random variable $X$, which follows a binomial distribution. This is denoted as:

$$ 
X \sim Binomial(n,p)
$$

This notation conveys that:

* The random variable $X$ takes values from a countably finite sample space ranging from $0$ to $n$, representing the possible number of successes.
* The **parameters** of the distribution are $n$, the number of independent trials, and $p$, the probability of success in each trial.

The probability mass function (pmf) for $X$ assigns probabilities to each possible value in this sample space, allowing us to compute probabilities, expected values, and other relevant quantities for the binomial experiment.

````python
x = range(0, 10)
p_x = binom.pmf(x, n=10, p=0.5)
plt.bar(x, p_x)
plt.show()
````

````python
# Generate 1000 random samples from a Binomial(n=10, p=0.5) using a Python list comprehension.
# Each sample is generated by calling rvs() individually. 
# (Less efficient because rvs() is called 1000 separate times.)
# Returns a Python list of length 1000.

x = [binom.rvs(n=10, p=0.5) for _ in range(1000)]
x
````

````python
# Generate 1000 random samples from a Binomial(n=10, p=0.5) distribution in a single call.
# size=1000 tells rvs to generate 1000 samples at once. Efficient because it’s vectorized.
# Returns a NumPy array of length 1000.

x = binom.rvs(n=10, p=0.5, size=1000)
x
````
